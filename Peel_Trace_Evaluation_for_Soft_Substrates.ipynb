```python
_t0 = step_timer_start("Step 3 mapping and trace-integrity review")
require_reference_validation()
#@title Step 3 — Scan and confirm sheet mapping, geometry, columns, units, and labels
#@markdown Review detected columns, units, include/exclude status, sheet-specific geometry, labels, and notes in one batch-capable editor.

# Rerunning this cell intentionally reopens mapping review. After confirmation, only the final banner is shown.
globals()["SCAN_CONFIRMED"] = False

if not globals().get("USER_FILE_CONFIRMED", False):
    waiting("Step 3 waiting for user-data file", "Return to Step 2 and confirm a readable data file first.")
else:
    if "CONFIG" not in globals():
        CONFIG = fp.build_manuscript_baseline_config(input_path=INPUT_PATH, output_root="Peel_Analysis_Output_v1_4_0_rc7", width_mm=WIDTH_MM, thickness_mm=THICKNESS_MM, bond_length_mm=BOND_LENGTH_MM)
        CONFIG.sheet_mapping_file = ""
    if "SCAN" not in globals() or not isinstance(SCAN, dict) or not Path(SCAN.get("sheet_mapping_file", "")).exists():
        info_card("Scanning workbook", "Detecting sheets, candidate force/displacement columns, units, labels, and geometry defaults. This does not calculate final metrics.")
        SCAN = fp.scan_workbook(CONFIG)
    CONFIG.sheet_mapping_file = SCAN["sheet_mapping_file"]
    SCAN_MAP = pd.read_csv(SCAN["sheet_mapping_file"])
    if "notes" not in SCAN_MAP.columns:
        SCAN_MAP["notes"] = ""
    if "include" in SCAN_MAP.columns:
        SCAN_MAP["include"] = SCAN_MAP["include"].astype(bool)

    if WIDGETS_AVAILABLE:
        out = widgets.Output()
        preview_out = widgets.Output()
        hist_out = widgets.Output()
        final_status = widgets.Output()
        integrity_out = widgets.Output()

        def save_map():
            SCAN_MAP.to_csv(SCAN["sheet_mapping_file"], index=False)
            CONFIG.sheet_mapping_file = SCAN["sheet_mapping_file"]

        def refresh_tables():
            with preview_out:
                clear_output(wait=True)
                if len(SCAN_CORRECTION_HISTORY) == 0:
                    display(HTML("<div class='peel-small-note'>No manual corrections applied yet. The detected mapping is ready for final confirmation if it looks correct.</div>"))
                else:
                    cols = [c for c in SCAN_MAP.columns if c in ["sheet","include","main_group","adhesive","replicate_id","force_column","displacement_column","force_unit","displacement_unit","width_mm","thickness_mm","bond_length_mm","notes"]]
                    display_wrapped_table(SCAN_MAP[cols], "Updated mapping preview", max_rows=12)
            with hist_out:
                clear_output(wait=True)
                if len(SCAN_CORRECTION_HISTORY) > 0:
                    display_wrapped_table(pd.DataFrame(SCAN_CORRECTION_HISTORY), "Correction history", max_rows=40)

        # Trace-integrity preflight is shown inside the collapsible editor area so it disappears after confirmation.
        try:
            TRACE_INPUTS_FOR_AUDIT = fp.build_trace_inputs(CONFIG, SCAN_MAP)
            TRACE_INTEGRITY_AUDIT = fp.trace_integrity_audit_from_traces(TRACE_INPUTS_FOR_AUDIT, CONFIG)
            tifail = int((TRACE_INTEGRITY_AUDIT.get("trace_integrity_status", pd.Series(dtype=str)) == "FAIL").sum()) if not TRACE_INTEGRITY_AUDIT.empty else 0
            tiwarn = int((TRACE_INTEGRITY_AUDIT.get("trace_integrity_status", pd.Series(dtype=str)) == "WARN").sum()) if not TRACE_INTEGRITY_AUDIT.empty else 0
            with integrity_out:
                clear_output(wait=True)
                if tifail:
                    bad_card("Trace-integrity failures detected", f"{tifail} sheet(s) have severe preflight issues. Exclude/correct them before analysis.")
                elif tiwarn:
                    display(card("Trace-integrity warnings detected", f"{tiwarn} sheet(s) have warnings such as large displacement gaps or missing values. Analysis can proceed after review.", THEME["orange"], THEME["amber_bg"]))
                else:
                    ok_card("Trace-integrity preflight passed", "No severe missing/discontinuous displacement issues were detected in included sheets.")
                show_detail_button("Show trace-integrity audit", lambda: TRACE_INTEGRITY_AUDIT, button_style="")
        except Exception as exc:
            with integrity_out:
                clear_output(wait=True)
                display(card("Trace-integrity preflight could not complete", f"{type(exc).__name__}: {exc}", THEME["orange"], THEME["amber_bg"]))

        # LEFT PANEL: Search and select sheets
        all_sheets = list(SCAN_MAP["sheet"])
        search_box = widgets.Text(placeholder="Search sheets...", description="Search:", layout=widgets.Layout(width="95%"))
        sheet_sel = widgets.SelectMultiple(options=all_sheets, description="", layout=widgets.Layout(width="95%", height="260px"))
        select_all_btn = widgets.Button(description="Select all filtered", layout=widgets.Layout(width="95%"))

        def filter_sheets(*args):
            q = search_box.value.lower()
            sheet_sel.options = [s for s in all_sheets if q in s.lower()]
        search_box.observe(filter_sheets, "value")

        def select_all(*args):
            sheet_sel.value = tuple(sheet_sel.options)
        select_all_btn.on_click(select_all)

        left_panel = widgets.VBox([search_box, select_all_btn, sheet_sel], layout=widgets.Layout(width="35%"))

        # RIGHT PANEL: Editor controls
        include_mode = widgets.Dropdown(options=[("leave unchanged", ""), ("Include selected sheets", "include"), ("Exclude selected sheets", "exclude")], description="Include:", layout=widgets.Layout(width="95%"))
        group_txt = widgets.Text(description="Group:", layout=widgets.Layout(width="45%"))
        adhesive_txt = widgets.Text(description="Adhesive:", layout=widgets.Layout(width="45%"))
        rep_txt = widgets.Text(description="Replicate:", layout=widgets.Layout(width="45%"))
        width_ft = widgets.FloatText(value=WIDTH_MM, description="w mm:", layout=widgets.Layout(width="45%"))
        thick_ft = widgets.FloatText(value=THICKNESS_MM, description="t mm:", layout=widgets.Layout(width="45%"))
        bond_ft = widgets.FloatText(value=BOND_LENGTH_MM, description="bond mm:", layout=widgets.Layout(width="45%"))
        force_dd = widgets.Dropdown(description="Force col:", layout=widgets.Layout(width="45%"))
        disp_dd = widgets.Dropdown(description="Disp col:", layout=widgets.Layout(width="45%"))
        force_unit = widgets.Dropdown(options=["leave unchanged","N","mN","lbf"], description="F unit:", layout=widgets.Layout(width="45%"))
        disp_unit = widgets.Dropdown(options=["leave unchanged","mm","cm","in","um"], description="D unit:", layout=widgets.Layout(width="45%"))
        notes_txt = widgets.Text(description="Notes:", layout=widgets.Layout(width="95%"))
        t_peel_alert = widgets.Output()

        def load_from_first_selected(*args):
            selected = list(sheet_sel.value)
            if not selected:
                return
            sheet = selected[0]
            row = SCAN_MAP.loc[SCAN_MAP["sheet"] == sheet].iloc[0]
            group_txt.value = str(row.get("main_group", ""))
            adhesive_txt.value = str(row.get("adhesive", ""))
            rep_txt.value = str(row.get("replicate_id", ""))
            notes_txt.value = str(row.get("notes", ""))
            width_ft.value = float(row.get("width_mm", WIDTH_MM))
            thick_ft.value = float(row.get("thickness_mm", THICKNESS_MM))
            bond_ft.value = float(row.get("bond_length_mm", BOND_LENGTH_MM))
            try:
                df = fp.read_full_table_for_scan(INPUT_PATH, sheet)
                cols = list(df.columns)
            except Exception:
                cols = [str(row.get("force_column", "")), str(row.get("displacement_column", ""))]
            force_dd.options = ["leave unchanged"] + cols
            disp_dd.options = ["leave unchanged"] + cols
            force_dd.value = row.get("force_column") if row.get("force_column") in cols and len(selected) == 1 else "leave unchanged"
            disp_dd.value = row.get("displacement_column") if row.get("displacement_column") in cols and len(selected) == 1 else "leave unchanged"
            force_unit.value = "leave unchanged"
            disp_unit.value = "leave unchanged"
            include_mode.value = ""
            with t_peel_alert:
                clear_output(wait=False)
                if "t-peel" in adhesive_txt.value.lower() or "t-peel" in notes_txt.value.lower():
                    display(card("T-peel geometry check", "T-peel displacement can differ from bonded-overlap length because the test displacement includes arm motion and deformation. Confirm <b>bond mm</b> is the intended bonded overlap for your analysis.", THEME["orange"], THEME["amber_bg"]))
        sheet_sel.observe(load_from_first_selected, names="value")
        load_from_first_selected()

        def record_change(sheet, field, old, new):
            if str(old) != str(new):
                SCAN_CORRECTION_HISTORY.append({"correction_no": len(SCAN_CORRECTION_HISTORY)+1, "sheet": sheet, "field": field, "old_value": old, "new_value": new})

        apply_btn = widgets.Button(description="Apply changes to selected sheets", button_style="warning", layout=widgets.Layout(width="300px"))
        confirm_btn = widgets.Button(description="Confirm all mapping decisions", button_style="primary", layout=widgets.Layout(width="300px"))

        def _apply_common(_):
            selected = list(sheet_sel.value)
            if not selected:
                status_in(out, card("No sheets selected", "Select one or more sheets from the list first.", THEME["orange"], THEME["amber_bg"]))
                return
            for sheet in selected:
                idx = SCAN_MAP.index[SCAN_MAP["sheet"] == sheet][0]
                vals = {
                    "width_mm": float(width_ft.value),
                    "thickness_mm": float(thick_ft.value),
                    "bond_length_mm": float(bond_ft.value),
                }
                if include_mode.value == "include": vals["include"] = True
                elif include_mode.value == "exclude": vals["include"] = False
                if group_txt.value.strip(): vals["main_group"] = group_txt.value.strip()
                if adhesive_txt.value.strip(): vals["adhesive"] = adhesive_txt.value.strip()
                if rep_txt.value.strip() and len(selected) == 1: vals["replicate_id"] = rep_txt.value.strip()
                if notes_txt.value.strip(): vals["notes"] = notes_txt.value.strip()
                if force_unit.value != "leave unchanged": vals["force_unit"] = force_unit.value
                if disp_unit.value != "leave unchanged": vals["displacement_unit"] = disp_unit.value
                if force_dd.value != "leave unchanged": vals["force_column"] = force_dd.value
                if disp_dd.value != "leave unchanged": vals["displacement_column"] = disp_dd.value
                for k, v in vals.items():
                    if k not in SCAN_MAP.columns:
                        SCAN_MAP[k] = ""
                    old = SCAN_MAP.loc[idx, k]
                    record_change(sheet, k, old, v)
                    SCAN_MAP.loc[idx, k] = v
            save_map()
            refresh_tables()
            status_in(out, card("Mapping update applied", f"Updated {len(selected)} sheet(s).", THEME["green"], THEME["green_bg"]))

        apply_btn.on_click(_apply_common)

        def on_confirm(_):
            global SCAN_CONFIRMED, MAPPING_CHANGED, RESULTS_STALE
            save_map()
            SCAN_CONFIRMED = True
            MAPPING_CHANGED = True
            RESULTS_STALE = True
            try:
                mapping_editor_box.layout.display = "none"
                with preview_out: clear_output(wait=False)
                with hist_out: clear_output(wait=False)
                with out: clear_output(wait=False)
                with integrity_out: clear_output(wait=False)
            except Exception:
                pass
            n_inc = int(SCAN_MAP.get("include", pd.Series(dtype=bool)).astype(bool).sum())
            n_exc = len(SCAN_MAP) - n_inc
            banner_html = final_banner(
                "Mapping confirmed",
                f"{n_inc} sheet(s) included, {n_exc} excluded.<br>Controls are collapsed. <b>To change mapping, rerun Step 3.</b><br>Next: run Step 4 to activate a profile.",
                "ok",
            ).data
            status_in(final_status, widgets.HTML(banner_html))

        confirm_btn.on_click(on_confirm)

        right_panel = widgets.VBox([
            include_mode,
            widgets.HBox([group_txt, adhesive_txt]),
            widgets.HBox([rep_txt, width_ft]),
            widgets.HBox([thick_ft, bond_ft]),
            widgets.HBox([force_dd, disp_dd]),
            widgets.HBox([force_unit, disp_unit]),
            notes_txt,
            widgets.HBox([apply_btn, help_widget("Button guide", "<b>Apply changes:</b> saves your typed values to the highlighted sheets.<br><b>Include/Exclude:</b> use the Include dropdown before applying.<br><b>Confirm all:</b> locks the mapping and collapses this panel. To revise, rerun Step 3.", width="280px")]),
            t_peel_alert,
        ], layout=widgets.Layout(width="65%", padding="0 0 0 15px", border_left="1px solid #ddd"))

        split_editor = widgets.HBox([left_panel, right_panel], layout=widgets.Layout(margin="10px 0", border="1px solid #cfd8dc", padding="15px", border_radius="8px"))
        header_html = widgets.HTML(f"""
        <div style='border-left:6px solid {THEME['blue']};background:{THEME['pale_blue']};border-radius:10px;padding:12px 16px;margin:10px 0;font-family:Arial, Helvetica, sans-serif;'>
          <h3 style='margin:0 0 8px 0;color:#263238'>Step 3 — Mapping review</h3>
          <div style='color:#37474f;font-size:14px;line-height:1.45'>Mapping file: <code>{SCAN['sheet_mapping_file']}</code><br>
          Use the editor below for one sheet or multiple sheets. Select multiple sheets using Ctrl/Cmd + click. Changes are applied only to selected sheets.</div>
        </div>
        """)
        mapping_editor_box = widgets.VBox([
            header_html,
            integrity_out,
            widgets.HTML("<b>Batch/sheet editor</b><br><span style='font-size:12px;color:#555'>Use the left panel to search/select sheets. Use the right panel to define columns, units, geometry, labels, notes, and include/exclude status.</span>"),
            split_editor,
            widgets.HBox([confirm_btn, help_widget("Confirmation", "Locks the mapping and collapses the editor. To remap or revise values, rerun this Step 3 cell.", width="260px")]),
            out,
            preview_out,
            hist_out,
        ])

        display(mapping_editor_box)
        refresh_tables()
        status_in(final_status, final_banner("Waiting for mapping confirmation", "Apply corrections if needed, then click <b>Confirm all mapping decisions</b>. Later steps skip until mapping is confirmed.", "warn"))
        display(final_status)
    else:
        CONFIG.sheet_mapping_file = SCAN["sheet_mapping_file"]
        SCAN_CONFIRMED = True
        MAPPING_CHANGED = True
        RESULTS_STALE = True
        ok_card("Mapping confirmed", "Widgets unavailable; detected mapping accepted directly. Run Step 4.")

step_timer_end("Step 3 mapping and trace-integrity review", _t0)
```

```python
_t0 = step_timer_start("Step 6 recovery and user-profile tools")
require_reference_validation()
#@title Step 6 — Optional recovery, sensitivity, and profile builder
#@markdown Use this after diagnostics to create/run conservative NON_MANUSCRIPT_MODIFIED recovery profiles. User benchmarks are handled in Step 7.

if "OUTPUTS" not in globals():
    waiting("Recovery/profile tools not available yet", "Run Step 5 first. Step 6 then uses diagnostics from the completed run to suggest global recovery settings.")
else:
    fail_count = int(OUTPUTS["readiness_after_run"].iloc[0].get("metric_fail_count", 0))
    review_count = int(OUTPUTS["readiness_after_run"].iloc[0].get("manual_review_required_count", 0))
    if fail_count or review_count:
        display(card("Recovery profile may help", f"Current run: {fail_count} failed trace(s), {review_count} trace(s) requiring manual review. Suggested settings are for feasibility/recovery, not for changing the manuscript baseline.", THEME["orange"], THEME["amber_bg"]))
    else:
        display(card("Sensitivity/profile mode", "All traces passed the current profile without quality warnings. A modified run is not needed for manuscript replication, but experienced users may save or test a documented user profile.", THEME["blue"], THEME["pale_blue"]))

    display(HTML("""
    <div style='font-family:Arial;margin:12px 0;padding:14px;border-left:6px solid #1177B2;background:#f4f8ff;border-radius:10px;line-height:1.45'>
    <b>How suggested settings are calculated:</b><br>
    Suggestions are based on trace feasibility and QC diagnostics, not adhesive ranking. The notebook checks usable displacement span after start/tail guards, whether the 25 mm manuscript window can fit, selected-window point count, drift ratio, peak/trough count, spatial spread of retained Top5/Bot5 extrema, and repeated warnings across sheets. The recovery window uses about 85% of the lower-quartile feasible travel, with a default floor of 12.5 mm (half of the manuscript 25 mm window). Anything below 12.5 mm requires advanced manual judgment.
    </div>
    """))
    param_effects = pd.DataFrame([
        {"parameter":"window_length_mm", "baseline":"25.0", "what_changing_does":"Shorter windows may fit short traces.", "risk_if_overused":"Too short can make metrics local and less representative."},
        {"parameter":"start_offset_mm", "baseline":"5.0", "what_changing_does":"Excludes toe/slack/startup displacement.", "risk_if_overused":"Too large may skip valid peel data."},
        {"parameter":"tail_exclude_frac", "baseline":"0.20", "what_changing_does":"Avoids terminal drop/failure region.", "risk_if_overused":"Too large leaves too little usable travel."},
        {"parameter":"max_window_drift_frac", "baseline":"0.25", "what_changing_does":"Rejects ramping/non-plateau windows.", "risk_if_overused":"Too loose accepts stretching-dominated regions; too strict may reject all windows."},
        {"parameter":"peak_prominence_frac", "baseline":"0.003 × Fmax", "what_changing_does":"Controls peak/trough sensitivity.", "risk_if_overused":"Too low detects noise; too high misses real extrema."},
        {"parameter":"strict_extrema_spread_check", "baseline":"advisory/off", "what_changing_does":"Converts clustered-extrema caution into failure.", "risk_if_overused":"May reject traces that still need human interpretation."},
    ])
    display_wrapped_table(param_effects, "Parameter-effect guide", max_rows=10)
    sugg = fp.suggest_global_settings_from_results(OUTPUTS["results"], CONFIG)
    if "suggested_global" in sugg.columns:
        current_profile_vals = {
            "window_length_mm": getattr(CONFIG, "window_length_mm", None),
            "start_offset_mm": getattr(CONFIG, "start_offset_mm", None),
            "tail_exclude_frac": getattr(CONFIG, "tail_exclude_frac", None),
            "max_window_drift_frac": getattr(CONFIG, "max_window_drift_frac", None),
            "peak_prominence_frac": getattr(CONFIG, "peak_prominence_frac", None),
            "strict_extrema_spread_check": getattr(CONFIG, "strict_extrema_spread_check", None),
        }
        sugg["current_profile"] = sugg["parameter"].map(current_profile_vals)
        sugg["user_edits_start_here"] = sugg["suggested_global"]
    changed_n = int(sugg.get("changed", pd.Series(dtype=bool)).fillna(False).astype(bool).sum()) if isinstance(sugg, pd.DataFrame) and "changed" in sugg.columns else 0
    if changed_n == 0:
        display(card("No recovery setting is recommended", "The current profile appears feasible for this dataset based on available diagnostics. A modified run is optional and mainly useful for sensitivity analysis.", THEME["blue"], THEME["pale_blue"]))
    display_wrapped_table(sugg, "Current profile vs suggested global recovery settings", max_rows=10)
    try:
        tmp_cfg = fp.build_manuscript_baseline_config(input_path=INPUT_PATH, output_root="_tmp", width_mm=WIDTH_MM, thickness_mm=THICKNESS_MM, bond_length_mm=BOND_LENGTH_MM)
        tmp_cfg.sheet_mapping_file = SCAN["sheet_mapping_file"]
        feas0 = fp.infeasible_sheets_from_mapping(pd.read_csv(SCAN["sheet_mapping_file"]), CONFIG)
        bad0 = feas0[feas0["status"].isin(["WINDOW_TOO_LONG_FOR_SHEET", "ADVANCED_SHORT_WINDOW"])] if isinstance(feas0, pd.DataFrame) and not feas0.empty else pd.DataFrame()
        if not bad0.empty:
            display_wrapped_table(bad0[[c for c in ["sheet","rough_displacement_span_mm","estimated_feasible_window_mm","active_window_length_mm","status","message"] if c in bad0.columns]], "Sheet-aware window feasibility for current profile", max_rows=15)
    except Exception as exc:
        display(card("Window feasibility preview unavailable", f"<code>{type(exc).__name__}: {exc}</code>", THEME["orange"], THEME["amber_bg"]))
    advice = fp.failed_trace_recovery_advice(OUTPUTS["results"])
    display_wrapped_table(advice, "Recovery advice", max_rows=10)

    if WIDGETS_AVAILABLE:
        def _suggest_value(param, fallback):
            try:
                return sugg[sugg["parameter"] == param].iloc[0]["suggested_global"]
            except Exception:
                return fallback
        profile_name = widgets.Text(value="user_profile_recovery_v1", description="Name:", layout=widgets.Layout(width="520px"))
        profile_note = widgets.Textarea(value="Recovery/sensitivity profile created from Step 6 diagnostics.", description="Note:", layout=widgets.Layout(width="95%", height="70px"))
        win = widgets.FloatText(value=float(_suggest_value("window_length_mm", 25.0)), description="window mm:", layout=widgets.Layout(width="260px"))
        start = widgets.FloatText(value=float(_suggest_value("start_offset_mm", 5.0)), description="start mm:", layout=widgets.Layout(width="260px"))
        tail = widgets.FloatText(value=float(_suggest_value("tail_exclude_frac", 0.20)), description="tail frac:", layout=widgets.Layout(width="260px"))
        drift = widgets.FloatText(value=float(_suggest_value("max_window_drift_frac", 0.25)), description="drift limit:", layout=widgets.Layout(width="260px"))
        prom = widgets.FloatText(value=float(_suggest_value("peak_prominence_frac", 0.003)), description="prom frac:", layout=widgets.Layout(width="260px"))
        strict_spread = widgets.Checkbox(value=False, description="strict extrema-spread QC")
        acknowledge = widgets.Checkbox(value=False, description="I understand this creates NON_MANUSCRIPT_MODIFIED outputs that should not be mixed with manuscript-baseline results.", layout=widgets.Layout(width="95%"))
        run_btn = widgets.Button(description="Run NON_MANUSCRIPT_MODIFIED profile", button_style="warning", layout=widgets.Layout(width="340px"))
        save_btn = widgets.Button(description="Save user profile JSON", button_style="", layout=widgets.Layout(width="240px"))
        reset_btn = widgets.Button(description="Reset to suggested values", button_style="primary", layout=widgets.Layout(width="230px"))
        out = widgets.Output()

        def set_suggested(_=None):
            win.value = float(_suggest_value("window_length_mm", 25.0)); start.value = float(_suggest_value("start_offset_mm", 5.0)); tail.value = float(_suggest_value("tail_exclude_frac", 0.20)); drift.value = float(_suggest_value("max_window_drift_frac", 0.25)); prom.value = float(_suggest_value("peak_prominence_frac", 0.003)); strict_spread.value = False
        reset_btn.on_click(set_suggested)
        def changed_from_baseline():
            return not (abs(float(win.value)-25.0)<1e-12 and abs(float(start.value)-5.0)<1e-12 and abs(float(tail.value)-0.20)<1e-12 and abs(float(drift.value)-0.25)<1e-12 and abs(float(prom.value)-0.003)<1e-12 and not bool(strict_spread.value))
        def validate_fields():
            if not acknowledge.value:
                return "Check the acknowledgment box before saving/running a NON_MANUSCRIPT_MODIFIED profile."
            if not changed_from_baseline():
                return "No settings changed from manuscript_baseline_v1. Use manuscript-baseline outputs instead, or intentionally change at least one setting."
            if float(win.value) <= 0 or float(start.value) < 0 or not (0 <= float(tail.value) < 0.8) or float(drift.value) <= 0 or float(prom.value) <= 0:
                return "Check that window > 0, start ≥ 0, 0 ≤ tail fraction < 0.8, drift limit > 0, and prominence fraction > 0."
            return ""
        def build_mod_cfg():
            cfg = fp.ManuscriptConfig(input_path=INPUT_PATH, output_root="Peel_Analysis_Output_v1_4_0_rc7_modified", width_mm=WIDTH_MM, thickness_mm=THICKNESS_MM, bond_length_mm=BOND_LENGTH_MM, assume_global_units=True, force_unit="N", displacement_unit="mm", start_offset_mm=float(start.value), window_length_mm=float(win.value), tail_exclude_frac=float(tail.value), max_window_drift_frac=float(drift.value), peak_prominence_frac=float(prom.value), min_peak_distance_points=1, min_window_points=10, required_top_peaks=5, required_bottom_troughs=5, break_proxy_consecutive_points=1, strict_extrema_spread_check=bool(strict_spread.value), export_qc_plots=True, export_group_summary=True, export_excel=True, make_zip=True, allow_non_manuscript_run=True, method_profile="non_manuscript_modified")
            cfg.sheet_mapping_file = SCAN["sheet_mapping_file"]
            return cfg
        def on_save(_):
            with out:
                clear_output(wait=True)
                msg = validate_fields()
                if msg:
                    display(card("User profile not saved", msg, THEME["red"], THEME["red_bg"])); return
                cfg = build_mod_cfg()
                name = profile_name.value.strip() or "user_profile_recovery_v1"
                path = f"{name}.json" if not name.endswith('.json') else name
                fp.save_method_profile_from_config(cfg, path, name)
                display(card("User profile saved", f"Saved <code>{path}</code>.<br>Note: {profile_note.value or 'none'}<br>Download it if using Binder/Colab because session files may not persist.", THEME["green"], THEME["green_bg"]))
        def on_run(_):
            global MODIFIED_OUTPUTS, RUN_HISTORY
            with out:
                clear_output(wait=True)
                msg = validate_fields()
                if msg:
                    display(card("Modified run not started", msg, THEME["red"], THEME["red_bg"])); return
                mod_cfg = build_mod_cfg()
                try:
                    feas = fp.infeasible_sheets_from_mapping(pd.read_csv(SCAN["sheet_mapping_file"]), mod_cfg)
                    bad = feas[feas["status"].isin(["WINDOW_TOO_LONG_FOR_SHEET"])] if isinstance(feas, pd.DataFrame) and not feas.empty else pd.DataFrame()
                    if not bad.empty:
                        display(card("Recovery window caution", f"{len(bad)} mapped sheet(s) may not support the selected recovery window. Exclude affected sheets in Step 3, shorten the window, or run manually with caution.", THEME["orange"], THEME["amber_bg"]))
                        display_wrapped_table(bad[[c for c in ["sheet","rough_displacement_span_mm","estimated_feasible_window_mm","active_window_length_mm","status","message"] if c in bad.columns]], "Sheets affected by recovery-window choice", max_rows=15)
                        return
                except Exception:
                    pass
                display(card("Running modified analysis", "This message will be replaced when the run is complete.", THEME["orange"], THEME["amber_bg"]))
                mod_fp = json.dumps({"file": globals().get("USER_FILE_INFO", {}).get("sha256", "")[:16], "mapping": SCAN.get("sheet_mapping_file", ""), "profile": {"window": float(win.value), "start": float(start.value), "tail": float(tail.value), "drift": float(drift.value), "prom": float(prom.value), "strict": bool(strict_spread.value)}, "backend": getattr(fp, "__version__", "")}, sort_keys=True)
                if mod_fp in globals().get("MODIFIED_RUN_FINGERPRINTS", set()):
                    clear_output(wait=True)
                    display(card("Duplicate modified run blocked", "This NON_MANUSCRIPT_MODIFIED parameter set has already been run for the current file/mapping/backend. Change at least one parameter, use previous outputs, or intentionally create a new profile name before rerunning.", THEME["orange"], THEME["amber_bg"]))
                    return
                MODIFIED_RUN_FINGERPRINTS = globals().setdefault("MODIFIED_RUN_FINGERPRINTS", set())
                MODIFIED_RUN_FINGERPRINTS.add(mod_fp)
                MODIFIED_OUTPUTS = fp.run_manuscript_pipeline(mod_cfg)
                clear_output(wait=True)
                run_label = f"Run {len(RUN_HISTORY)+1}: NON_MANUSCRIPT_MODIFIED"
                RUN_HISTORY.append(fp.run_record_from_outputs(MODIFIED_OUTPUTS, mod_cfg, run_label=run_label))
                display(card("Modified analysis complete", f"{run_label}<br>Run folder: <code>{MODIFIED_OUTPUTS['outdir']}</code><br>Main ZIP: <code>{MODIFIED_OUTPUTS['zip_file']}</code>", THEME["green"], THEME["green_bg"]))
                display_wrapped_table(MODIFIED_OUTPUTS.get("readiness_after_run", pd.DataFrame()), "Modified-run compact status", max_rows=5)
                display_wrapped_table(pd.DataFrame(RUN_HISTORY), "Updated run comparison", max_rows=20)
        save_btn.on_click(on_save); run_btn.on_click(on_run)
        display(widgets.VBox([
            profile_name, profile_note, widgets.HBox([reset_btn]),
            widgets.HBox([win, help_widget("Window length", "Selected displacement span. If unchanged from an already-run profile, the duplicate-run guard will prevent wasting time.", width="260px"), start, help_widget("Start offset", "Initial displacement excluded before window search.", width="240px")]),
            widgets.HBox([tail, help_widget("Tail exclusion", "Terminal fraction excluded from window search.", width="240px"), drift, help_widget("Drift limit", "Window ramp/plateau guard: |slope|×window length/mean force.", width="260px")]),
            widgets.HBox([prom, help_widget("Peak prominence", "Peak/trough threshold as a fraction of global corrected force maximum.", width="260px"), strict_spread, help_widget("Strict extrema-spread QC", "Treats clustered extrema as failures rather than advisory review flags.", width="280px")]),
            acknowledge, widgets.HBox([run_btn, save_btn]), out
        ]))
    else:
        waiting("Widgets unavailable", "Create a user profile manually with fp.ManuscriptConfig(..., allow_non_manuscript_run=True).")

step_timer_end("Step 6 recovery and user-profile tools", _t0)
```

## Interpretation notes

- `SSA` is an operational force-oscillation descriptor under the same machine/protocol, not an intrinsic crack-front material property.
- `PSI` is only meaningful when the selected window is physically plausible and passes the drift/tail/window checks.
- `MANUSCRIPT_BASELINE` outputs reproduce the locked analysis settings. `NON_MANUSCRIPT_MODIFIED` outputs are recovery/sensitivity runs and should not be mixed with manuscript-baseline results without explicit disclosure.
- `displacement_at_break_mm` uses a post-window `0.10 × Fc` drop proxy when found; if no drop is detected, the final recorded displacement is used and flagged.
- QC plots use color-blind-safe, paper-matched colors, but interpretation should rely on both color and marker/line style plus the exported diagnostic tables.
