```python
#@title Step 0 â€” Setup: imports, display utilities, runtime detection, and backend loading
#@markdown This cell loads the v1.4.0-rc15 analysis engine, notebook styling, and reusable UI helpers. It does not analyze user data.

import os, sys, json, time, shutil, gc, platform
from pathlib import Path
import pandas as pd
from IPython.display import display, HTML, Markdown, clear_output

try:
    import ipywidgets as widgets
    WIDGETS_AVAILABLE = True
except Exception:
    WIDGETS_AVAILABLE = False

IN_COLAB = "google.colab" in sys.modules
sys.path.insert(0, os.getcwd())
sys.path.insert(0, "/mnt/data")

print("Importing backend module: fabric_peel_guided_core_v1_4_0_rc15.py")
import fabric_peel_guided_core_v1_4_0_rc15 as fp
print("Backend imported successfully. Version:", getattr(fp, "__version__", "unknown"))
fp.apply_paper_plot_style()

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", None)
pd.set_option("display.width", 220)

display(HTML("""
<style>
.dataframe {font-family: Arial, Helvetica, sans-serif !important; font-size: 12.5px !important; border-collapse: collapse !important;}
.dataframe td {white-space: normal !important; word-wrap: break-word !important; max-width: 280px !important; text-align: left !important; vertical-align: top !important; padding: 6px !important;}
.dataframe th {white-space: normal !important; text-align: left !important; vertical-align: top !important; padding: 6px !important; background:#eef3f7 !important;}
.peel-small-note {font-family: Arial, Helvetica, sans-serif; font-size: 12.5px; color: #455a64; line-height: 1.45;}
.peel-code {font-family: Consolas, Menlo, monospace; background:#f3f5f6; padding:1px 4px; border-radius:4px;}
</style>
"""))

THEME = {
    "blue": "#1177B2", "light_blue": "#5AA7D6", "gray": "#8C8C8C", "orange": "#DD6B00",
    "black": "black", "green": "#2e7d32", "red": "#b71c1c", "pale_blue": "#f4f8ff",
    "amber_bg": "#fff8e1", "red_bg": "#fff5f5", "green_bg": "#f4fbf5", "gray_bg": "#f7f9fa"
}

def card(title, body, color="#546e7a", background="#f8fafb"):
    return HTML(f"""
    <div style="border-left:6px solid {color};background:{background};border-radius:10px;padding:12px 16px;margin:10px 0;font-family:Arial, Helvetica, sans-serif;max-width:100%">
      <h3 style="margin:0 0 8px 0;color:#263238">{title}</h3>
      <div style="color:#37474f;font-size:14px;line-height:1.45">{body}</div>
    </div>
    """)

def waiting(title, body="Complete the prior checkpoint, then run this cell again."):
    display(card(title, body, THEME["orange"], THEME["amber_bg"]))

def ok_card(title, body):
    display(card(title, body, THEME["green"], THEME["green_bg"]))

def bad_card(title, body):
    display(card(title, body, THEME["red"], THEME["red_bg"]))

def info_card(title, body):
    display(card(title, body, THEME["blue"], THEME["pale_blue"]))

def status_in(out, obj):
    """Clear one shared output area and display a card/table there."""
    with out:
        clear_output(wait=True)
        display(obj)

def status_color(status):
    s = str(status).upper()
    if "FAIL" in s:
        return THEME["red"], THEME["red_bg"]
    if "CAUTION" in s or "REVIEW" in s or "WARNING" in s:
        return THEME["orange"], THEME["amber_bg"]
    if "PASS" in s or "OK" in s or "QUALITY" in s:
        return THEME["green"], THEME["green_bg"]
    return THEME["blue"], THEME["pale_blue"]

def display_wrapped_table(df, title=None, max_rows=25):
    if title:
        display(HTML(f"<h4 style='font-family:Arial;margin:8px 0 4px 0;color:#263238'>{title}</h4>"))
    if df is None or len(df) == 0:
        display(HTML("<div class='peel-small-note'>No rows to display.</div>"))
        return
    show = df.head(max_rows).copy()
    display(show)
    if len(df) > max_rows:
        display(HTML(f"<div class='peel-small-note'>Detailed audit table is truncated in this notebook view: showing first {max_rows} of {len(df)} rows. The full audit is saved in the exported CSV/Excel files.</div>"))

def display_status_legend():
    display(HTML("""
    <div style='font-family:Arial;margin:8px 0;padding:10px;border-radius:8px;background:#f7f9fa;border-left:5px solid #8C8C8C'>
    <b>Status color legend:</b>
    <span style='background:#f4fbf5;border-left:4px solid #2e7d32;padding:3px 8px;margin-left:8px'>Green = pass / quality OK</span>
    <span style='background:#fff8e1;border-left:4px solid #DD6B00;padding:3px 8px;margin-left:8px'>Amber = computed but review required</span>
    <span style='background:#fff5f5;border-left:4px solid #b71c1c;padding:3px 8px;margin-left:8px'>Red = fail / do not interpret</span>
    <span style='background:#f4f8ff;border-left:4px solid #1177B2;padding:3px 8px;margin-left:8px'>Blue = information / baseline</span>
    </div>
    """))

def normalize_upload_value(value):
    if not value:
        return []
    if isinstance(value, dict):
        return [{"name": k, "content": v.get("content", b""), "size": len(v.get("content", b""))} for k, v in value.items()]
    if isinstance(value, (tuple, list)):
        return list(value)
    return []

def format_size(nbytes):
    try:
        n = float(nbytes)
    except Exception:
        return "unknown size"
    for unit in ["B", "KB", "MB", "GB"]:
        if n < 1024:
            return f"{n:.1f} {unit}"
        n /= 1024
    return f"{n:.1f} TB"

def write_json(obj, path):
    Path(path).write_text(json.dumps(obj, indent=2), encoding='utf-8')
    return str(Path(path).resolve())

# Preserve state across reruns, when appropriate.
INPUT_PATH = globals().get("INPUT_PATH", "")
USER_FILE_CONFIRMED = globals().get("USER_FILE_CONFIRMED", False)
USER_FILE_INFO = globals().get("USER_FILE_INFO", {})
WIDTH_MM = float(globals().get("WIDTH_MM", 20.0))
THICKNESS_MM = float(globals().get("THICKNESS_MM", 0.40))
BOND_LENGTH_MM = float(globals().get("BOND_LENGTH_MM", 25.0))
REFERENCE_VALIDATION_PASSED = globals().get("REFERENCE_VALIDATION_PASSED", False)
FILE_CHANGED = globals().get("FILE_CHANGED", False)
PROFILE_CHANGED = globals().get("PROFILE_CHANGED", False)
MAPPING_CHANGED = globals().get("MAPPING_CHANGED", False)
RESULTS_STALE = globals().get("RESULTS_STALE", False)
SETTINGS_CONFIRMED = globals().get("SETTINGS_CONFIRMED", False)
SCAN_CONFIRMED = globals().get("SCAN_CONFIRMED", False)
RUN_HISTORY = globals().get("RUN_HISTORY", [])
MODIFIED_RUN_COUNTER = globals().get("MODIFIED_RUN_COUNTER", 1)
SCAN_CORRECTION_HISTORY = globals().get("SCAN_CORRECTION_HISTORY", [])
USER_PROFILE_PATH = globals().get("USER_PROFILE_PATH", "")
ACTIVE_PROFILE_KIND = globals().get("ACTIVE_PROFILE_KIND", "")
USER_BENCHMARK_EXPECTED_PATH = globals().get("USER_BENCHMARK_EXPECTED_PATH", "")


def step_timer_start(label):
    print(f"[{label}] started: {time.strftime('%Y-%m-%d %H:%M:%S')}")
    return time.time()

def step_timer_end(label, t0):
    print(f"[{label}] finished: {time.strftime('%Y-%m-%d %H:%M:%S')} | elapsed: {time.time()-t0:.2f} s")

def require_reference_validation():
    if not globals().get("REFERENCE_VALIDATION_PASSED", False):
        raise RuntimeError("Built-in reference validation has not passed. Run Step 1 first; if it fails, do not analyze user data until the package/reference validation is resolved.")

def mark_results_stale(reason):
    global RESULTS_STALE
    RESULTS_STALE = True
    display(card("Previous results marked stale", reason, THEME["orange"], THEME["amber_bg"]))

def _sha256_file16(path):
    """
    Return the first 16 characters of a file SHA-256 hash.

    Used for duplicate-run detection so that editing the mapping CSV or
    user-profile JSON changes the run fingerprint even if the file path is unchanged.
    """
    try:
        if not path:
            return ""

        p = Path(path)

        if not p.exists() or not p.is_file():
            return ""

        import hashlib

        h = hashlib.sha256()

        with p.open("rb") as f:
            for chunk in iter(lambda: f.read(1024 * 1024), b""):
                h.update(chunk)

        return h.hexdigest()[:16]

    except Exception:
        return ""


def _cfg_attr(cfg, name, default=""):
    """
    Safely read a CONFIG attribute for fingerprinting.
    """
    try:
        return getattr(cfg, name, default)
    except Exception:
        return default


def _profile_parameter_fingerprint(cfg):
    """
    Computational profile parameters that should trigger a new analysis run
    when changed.
    """
    if cfg is None:
        return {}

    return {
        "method_profile": _cfg_attr(cfg, "method_profile", ""),
        "allow_non_manuscript_run": bool(_cfg_attr(cfg, "allow_non_manuscript_run", False)),
        "window_length_mm": _cfg_attr(cfg, "window_length_mm", ""),
        "start_offset_mm": _cfg_attr(cfg, "start_offset_mm", ""),
        "tail_exclude_frac": _cfg_attr(cfg, "tail_exclude_frac", ""),
        "max_window_drift_frac": _cfg_attr(cfg, "max_window_drift_frac", ""),
        "peak_prominence_frac": _cfg_attr(cfg, "peak_prominence_frac", ""),
        "strict_extrema_spread_check": bool(_cfg_attr(cfg, "strict_extrema_spread_check", False)),
        "output_root": _cfg_attr(cfg, "output_root", ""),
    }


def state_fingerprint():
    """
    Fingerprint the active analysis state for duplicate-run blocking.

    This intentionally includes file CONTENT hashes, not only file paths.
    That prevents false duplicate blocking when:
      1. the Step 3 mapping CSV is edited but keeps the same path;
      2. a user-profile JSON is edited but keeps the same path;
      3. profile parameters are changed inside CONFIG.
    """
    cfg = globals().get("CONFIG", None)

    scan_obj = globals().get("SCAN", {})
    scan_mapping_file = ""

    if isinstance(scan_obj, dict):
        scan_mapping_file = scan_obj.get("sheet_mapping_file", "")

    config_mapping_file = _cfg_attr(cfg, "sheet_mapping_file", "")
    mapping_file = config_mapping_file or scan_mapping_file

    user_profile_path = globals().get("USER_PROFILE_PATH", "")
    input_path = globals().get("INPUT_PATH", "")

    user_file_info = globals().get("USER_FILE_INFO", {})
    stored_input_sha = ""

    if isinstance(user_file_info, dict):
        stored_input_sha = str(user_file_info.get("sha256", ""))

    input_sha16 = stored_input_sha[:16] if stored_input_sha else _sha256_file16(input_path)

    return {
        "input_path": str(input_path),
        "input_sha16": input_sha16,

        "mapping_file": str(mapping_file),
        "mapping_file_sha16": _sha256_file16(mapping_file),

        "active_profile_kind": globals().get("ACTIVE_PROFILE_KIND", ""),
        "active_profile_note": globals().get("ACTIVE_PROFILE_NOTE", ""),

        "user_profile_path": str(user_profile_path),
        "user_profile_sha16": _sha256_file16(user_profile_path),

        "profile_parameters": _profile_parameter_fingerprint(cfg),

        "backend_version": getattr(fp, "__version__", "unknown"),
    }

ok_card(
    "Backend loaded",
    "<b>Module:</b> fabric_peel_guided_core_v1_4_0_rc15<br>"
    "<b>Built-in method:</b> manuscript_baseline_v1 for strict Top5+Bottom5 force-trace extraction<br>"
    "<b>User flexibility:</b> saved user profiles and user benchmarks are supported without overwriting the manuscript baseline<br>"
    "<b>Interface:</b> non-blocking callbacks + guarded downstream cells<br>"
    "<b>v1.4.0-rc15 polish:</b> collapsed confirmed panels, sheet-aware window feasibility, compact help panels, duplicate-run blocking, and batch include/exclude editing overrides, improved rerun map, and clearer validation-count language"
)
display_status_legend()


# Central lightweight state dictionary for display logic. The notebook still stores key
# variables as globals for compatibility with widgets, but panels should read this summary.
WORKFLOW_STATE = globals().get("WORKFLOW_STATE", {
    "active_file": globals().get("INPUT_PATH", ""),
    "active_profile": globals().get("ACTIVE_PROFILE_KIND", ""),
    "profile_note": globals().get("ACTIVE_PROFILE_NOTE", ""),
    "results_stale": globals().get("RESULTS_STALE", False),
})

def update_workflow_state(**kwargs):
    WORKFLOW_STATE.update(kwargs)
    return WORKFLOW_STATE

def compact_status_table(rows, title="Status summary"):
    df = pd.DataFrame(rows)
    display_wrapped_table(df, title, max_rows=30)

def show_detail_button(label, producer, button_style=""):
    """Display a button that renders a detailed table only on request, with a close button."""
    if not WIDGETS_AVAILABLE:
        try:
            display_wrapped_table(producer(), label)
        except Exception as exc:
            display(card(f"{label} unavailable", f"<code>{type(exc).__name__}: {exc}</code>", THEME["orange"], THEME["amber_bg"]))
        return

    out = widgets.Output()
    btn = widgets.Button(description=label, button_style=button_style, layout=widgets.Layout(width="280px"))
    close_btn = widgets.Button(description="Close table", layout=widgets.Layout(width="120px"))

    def _on_click(_):
        with out:
            clear_output(wait=True)
            try:
                obj = producer()
                if isinstance(obj, pd.DataFrame):
                    display_wrapped_table(obj, label, max_rows=80)
                else:
                    display(obj)
                # Show the close button at the bottom of the table
                display(close_btn)
            except Exception as exc:
                display(card(f"{label} unavailable", f"<code>{type(exc).__name__}: {exc}</code>", THEME["orange"], THEME["amber_bg"]))
                display(close_btn)

    def _hide(_):
        with out:
            # wait=False instantly clears the table and the close button
            clear_output(wait=False)

    btn.on_click(_on_click)
    close_btn.on_click(_hide)

    display(widgets.VBox([btn, out]))

def show_detail_button_with_help(label, producer, help_title, help_body, button_style=""):
    """Display an on-demand detail table button with a separate closable ? help widget."""
    if not WIDGETS_AVAILABLE:
        show_detail_button(label, producer, button_style=button_style)
        return
    out = widgets.Output()
    btn = widgets.Button(description=label, button_style=button_style, layout=widgets.Layout(width="280px"))
    close_btn = widgets.Button(description="Close table", layout=widgets.Layout(width="120px"))
    def _on_click(_):
        with out:
            clear_output(wait=True)
            try:
                obj = producer()
                if isinstance(obj, pd.DataFrame):
                    display_wrapped_table(obj, label, max_rows=80)
                else:
                    display(obj)
                display(close_btn)
            except Exception as exc:
                display(card(f"{label} unavailable", f"<code>{type(exc).__name__}: {exc}</code>", THEME["orange"], THEME["amber_bg"]))
                display(close_btn)
    def _hide(_):
        with out:
            clear_output(wait=False)
    btn.on_click(_on_click)
    close_btn.on_click(_hide)
    display(widgets.VBox([widgets.HBox([btn, help_widget(help_title, help_body, width="360px")]), out]))


# Compact, closable help panels. Use these instead of permanent long explanations.
def help_widget(title, body, width="95%"):
    """Small closable help widget. Pure ipywidgets; no HTML/JS IDs."""
    if not WIDGETS_AVAILABLE:
        return HTML(f"<div class='peel-small-note'><b>{title}</b><br>{body}</div>")
    out = widgets.Output()
    btn = widgets.Button(description="?", tooltip=title, layout=widgets.Layout(width="34px"), button_style="")
    close = widgets.Button(description="Close", layout=widgets.Layout(width="80px"), button_style="")
    close.layout.display = "none"
    def show(_):
        close.layout.display = "inline-flex"
        with out:
            clear_output(wait=True)
            display(card(title, body, THEME["blue"], THEME["pale_blue"]))
    def hide(_):
        close.layout.display = "none"
        with out:
            clear_output(wait=False)
    btn.on_click(show)
    close.on_click(hide)
    return widgets.VBox([widgets.HBox([btn, close]), out], layout=widgets.Layout(width=width))

def profile_file_candidates(folder="."):
    """Find saved user profiles, excluding templates, manuscript baselines, and benchmark metadata."""
    pats = ["user_profile_*.json", "user_method_profile_*.json", "profile_*.json"]
    found = []
    for pat in pats:
        found.extend(Path(folder).glob(pat))
    out = []
    for p in found:
        name = p.name.lower()
        if ("manuscript_baseline" in name or "template" in name or "benchmark_metadata" in name or
            "user_benchmark" in name or "expected" in name or "scotch" in name):
            continue
        if str(p) not in out:
            out.append(str(p))
    return sorted(out)

def user_benchmark_candidates(folder="."):
    found = []
    for pat in ["user_benchmark_expected*.csv", "user_benchmark_*_expected.csv", "*_benchmark_expected.csv"]:
        found.extend(Path(folder).glob(pat))
    out=[]
    for p in found:
        name=p.name.lower()
        if "scotch" in name or "template" in name:
            continue
        if str(p) not in out:
            out.append(str(p))
    return sorted(out)

def safe_profile_id(text):
    import re
    s = re.sub(r"[^a-zA-Z0-9_-]+", "_", str(text).strip()).strip("_").lower()
    # Users type only a short ID.  Saved files are automatically named user_profile_<id>.json.
    # Strip common filename prefixes so future sessions can still auto-detect profiles cleanly.
    for prefix in ["user_method_profile_", "user_profile_", "profile_"]:
        if s.startswith(prefix):
            s = s[len(prefix):]
    s = s.strip("_")
    if not s:
        s = "custom_profile"
    return s[:32]


# rc6: final-status banners, active-state strip, and robust file/path helpers.
def final_banner(title, body, status="ok"):
    color, bg = {
        "ok": (THEME["green"], THEME["green_bg"]),
        "info": (THEME["blue"], THEME["pale_blue"]),
        "warn": (THEME["orange"], THEME["amber_bg"]),
        "fail": (THEME["red"], THEME["red_bg"]),
    }.get(status, (THEME["blue"], THEME["pale_blue"]))
    return card(title, body, color, bg)

def display_active_state():
    rows = [{
        "active_file": Path(globals().get("INPUT_PATH", "")).name if globals().get("INPUT_PATH", "") else "not selected",
        "mapping": "confirmed" if globals().get("SCAN_CONFIRMED", False) else "not confirmed",
        "active_profile": globals().get("ACTIVE_PROFILE_KIND", "not selected") or "not selected",
        "latest_run": len(globals().get("RUN_HISTORY", [])),
        "results_stale": globals().get("RESULTS_STALE", False),
    }]
    display_wrapped_table(pd.DataFrame(rows), "Current workflow state", max_rows=5)

def is_allowed_data_file(path):
    p = Path(path)
    if p.suffix.lower() not in {".xlsx", ".xlsm", ".xls", ".csv"}:
        return False
    low = p.name.lower()
    excluded_terms = ["benchmark", "profile", "method_profile", "template", "validation", "readme", "requirements", "run_history", "diagnostic", "trace_integrity", "output_consistency", "citation", "license"]
    return not any(term in low for term in excluded_terms)

def large_file_note(path):
    try:
        size = Path(path).stat().st_size
    except Exception:
        return ""
    if size > 100*1024*1024:
        return "<br><b>Large-file note:</b> file is >100 MB. Google Drive/manual path is strongly recommended in Colab."
    if size > 50*1024*1024:
        return "<br><b>Large-file note:</b> file is >50 MB. Use Google Drive/manual path if browser upload is slow."
    return ""
```

```python
_t0 = step_timer_start("Step 3 mapping and trace-integrity review")
require_reference_validation()

#@title Step 3 â€” Scan and confirm sheet mapping, geometry, columns, units, and labels
#@markdown Review detected columns, units, include/exclude status, sheet-specific geometry, labels, and notes in the split-panel batch editor.

# Rerunning this cell intentionally reopens mapping review. After confirmation, only the final banner is shown.
globals()["SCAN_CONFIRMED"] = False


# ============================================================
# Step 3 helper functions
# ============================================================

_UNKNOWN_TOKENS = {"", "nan", "none", "unknown", "na", "n/a", "null", "<na>"}
_VALID_FORCE_UNITS = {"N", "mN", "lbf"}
_VALID_DISPLACEMENT_UNITS = {"mm", "cm", "in", "um"}

def _is_missing_like(value):
    if value is None:
        return True
    try:
        if pd.isna(value):
            return True
    except Exception:
        pass
    return str(value).strip().lower() in _UNKNOWN_TOKENS

def _clean_unit(value):
    """
    Normalize common unit spellings from the mapping CSV.
    Returns canonical units used by the analysis workflow.
    """
    if _is_missing_like(value):
        return ""

    s = str(value).strip()
    sl = s.lower().replace(" ", "")

    aliases = {
        # force
        "n": "N",
        "newton": "N",
        "newtons": "N",
        "mn": "mN",
        "millinewton": "mN",
        "millinewtons": "mN",
        "lbf": "lbf",
        "lb": "lbf",
        "lbs": "lbf",
        "pound": "lbf",
        "pounds": "lbf",
        "poundforce": "lbf",
        "poundsforce": "lbf",

        # displacement
        "mm": "mm",
        "millimeter": "mm",
        "millimeters": "mm",
        "cm": "cm",
        "centimeter": "cm",
        "centimeters": "cm",
        "in": "in",
        "inch": "in",
        "inches": "in",
        "um": "um",
        "Âµm": "um",
        "Î¼m": "um",
        "micron": "um",
        "microns": "um",
        "micrometer": "um",
        "micrometers": "um",
    }

    return aliases.get(sl, s)

def _valid_force_unit(value):
    return _clean_unit(value) in _VALID_FORCE_UNITS

def _valid_displacement_unit(value):
    return _clean_unit(value) in _VALID_DISPLACEMENT_UNITS



def _as_include_bool(value):
    if isinstance(value, bool):
        return value
    s = str(value).strip().lower()
    if s in {"false", "0", "no", "n", "exclude", "excluded"}:
        return False
    if s in {"true", "1", "yes", "y", "include", "included"}:
        return True
    return True

def _included_mask(df):
    if "include" in df.columns:
        return df["include"].apply(_as_include_bool).astype(bool)
    return pd.Series([True] * len(df), index=df.index)

def _clean_col(value):
    return "" if _is_missing_like(value) else str(value).strip()

def _is_positive_number(value):
    x = pd.to_numeric(value, errors="coerce")
    return pd.notna(x) and float(x) > 0

def _safe_positive_float(value, default):
    x = pd.to_numeric(value, errors="coerce")
    if pd.notna(x) and float(x) > 0:
        return float(x)
    return float(default)

def _trace_status_by_sheet(tia):
    if (
        isinstance(tia, pd.DataFrame)
        and not tia.empty
        and "sheet" in tia.columns
        and "trace_integrity_status" in tia.columns
    ):
        return dict(zip(tia["sheet"].astype(str), tia["trace_integrity_status"].astype(str)))
    return {}

def mapping_issue_masks(df, tia=None):
    if df is None or df.empty:
        empty = pd.Series(dtype=bool)
        return {
            k: empty for k in [
                "all",
                "included",
                "excluded",
                "low_conf",
                "trace_issue",
                "unknown",
                "missing_required",
                "invalid_geometry",
                "invalid_units",
                "identical_cols",
                "needs_review",
            ]
        }

    idx = df.index
    included = _included_mask(df)
    excluded = ~included

    # Low-confidence force/displacement assignment
    low_conf = pd.Series(False, index=idx)
    if "column_confidence" in df.columns:
        low_conf = ~df["column_confidence"].fillna("").astype(str).str.upper().isin(
            ["HIGH", "OK", "PASS"]
        )

    # Missing labels or required columns
    missing_required = pd.Series(False, index=idx)
    for col in ["main_group", "adhesive", "force_column", "displacement_column"]:
        if col in df.columns:
            missing_required |= df[col].apply(_is_missing_like)
        else:
            missing_required |= True

    # Missing, zero, or negative geometry
    invalid_geometry = pd.Series(False, index=idx)
    for col in ["width_mm", "thickness_mm", "bond_length_mm"]:
        if col in df.columns:
            vals = pd.to_numeric(df[col], errors="coerce")
            invalid_geometry |= vals.isna() | (vals <= 0)
        else:
            invalid_geometry |= True

    # Missing or unsupported units
    invalid_units = pd.Series(False, index=idx)

    if "force_unit" in df.columns:
        invalid_units |= ~df["force_unit"].apply(_valid_force_unit)
    else:
        invalid_units |= True

    if "displacement_unit" in df.columns:
        invalid_units |= ~df["displacement_unit"].apply(_valid_displacement_unit)
    else:
        invalid_units |= True

    # Force and displacement columns should not be the same column
    identical_cols = pd.Series(False, index=idx)
    if "force_column" in df.columns and "displacement_column" in df.columns:
        f_cols = df["force_column"].apply(_clean_col)
        d_cols = df["displacement_column"].apply(_clean_col)
        identical_cols = (f_cols != "") & (d_cols != "") & (f_cols == d_cols)

    # Scanner warnings/problem fields
    problem = pd.Series(False, index=idx)
    for col in ["problem", "warnings"]:
        if col in df.columns:
            problem |= ~df[col].fillna("").astype(str).str.strip().isin(["", "nan", "None"])

    # Trace-integrity preflight status
    trace_issue = pd.Series(False, index=idx)
    status_map = _trace_status_by_sheet(tia)
    if status_map and "sheet" in df.columns:
        trace_issue = df["sheet"].astype(str).map(status_map).fillna("").isin(["WARN", "FAIL"])

    unknown = missing_required | invalid_geometry | invalid_units

    needs_review = included & (
        low_conf
        | missing_required
        | invalid_geometry
        | invalid_units
        | identical_cols
        | problem
        | trace_issue
    )

    return {
        "all": pd.Series(True, index=idx),
        "included": included,
        "excluded": excluded,
        "low_conf": included & low_conf,
        "trace_issue": included & trace_issue,
        "unknown": included & unknown,
        "missing_required": included & missing_required,
        "invalid_geometry": included & invalid_geometry,
        "invalid_units": included & invalid_units,
        "identical_cols": included & identical_cols,
        "needs_review": needs_review,
    }

def mapping_precheck_banner(df, tia=None):
    masks = mapping_issue_masks(df, tia)

    total = int(len(df)) if df is not None else 0
    needs = int(masks["needs_review"].sum()) if total else 0
    low = int(masks["low_conf"].sum()) if total else 0
    trace = int(masks["trace_issue"].sum()) if total else 0
    missing = int(masks["missing_required"].sum()) if total else 0
    bad_geom = int(masks["invalid_geometry"].sum()) if total else 0
    bad_units = int(masks["invalid_units"].sum()) if total else 0
    dup_cols = int(masks["identical_cols"].sum()) if total else 0

    if needs:
        body = (
            f"{needs} of {total} sheet(s) need review:<br>"
            f"â€¢ {low} low-confidence force/displacement assignment(s)<br>"
            f"â€¢ {trace} trace-integrity warning/failure row(s)<br>"
            f"â€¢ {missing} missing label or required-column row(s)<br>"
            f"â€¢ {bad_geom} missing, zero, or negative geometry row(s)<br>"
            f"â€¢ {bad_units} missing or unsupported unit row(s)<br>"
            f"â€¢ {dup_cols} identical force/displacement column row(s)<br><br>"
            "Use the filters below to inspect and resolve these issues."
        )

        return card(
            "Mapping review needed",
            body,
            THEME["orange"],
            THEME["amber_bg"],
        )

    return card(
        "Mapping precheck passed",
        (
            f"All {total} sheet(s) have required columns, units, geometry, labels, "
            "and suitable trace-integrity preflight status for analysis."
        ),
        THEME["green"],
        THEME["green_bg"],
    )

def _display_cols(df):
    return [
        c for c in df.columns
        if c in [
            "sheet",
            "include",
            "main_group",
            "adhesive",
            "replicate_id",
            "force_column",
            "displacement_column",
            "force_unit",
            "displacement_unit",
            "width_mm",
            "thickness_mm",
            "bond_length_mm",
            "column_confidence",
            "warnings",
            "problem",
            "notes",
        ]
    ]


# ============================================================
# Main Step 3 logic
# ============================================================

if not globals().get("USER_FILE_CONFIRMED", False):
    waiting(
        "Step 3 waiting for user-data file",
        "Return to Step 2 and confirm a readable data file first.",
    )

else:
    if "CONFIG" not in globals():
        CONFIG = fp.build_manuscript_baseline_config(
            input_path=INPUT_PATH,
            output_root="Peel_Analysis_Output_v1_4_0_rc15",
            width_mm=WIDTH_MM,
            thickness_mm=THICKNESS_MM,
            bond_length_mm=BOND_LENGTH_MM,
        )
        CONFIG.sheet_mapping_file = ""

    if (
        "SCAN" not in globals()
        or not isinstance(SCAN, dict)
        or not Path(SCAN.get("sheet_mapping_file", "")).exists()
    ):
        info_card(
            "Scanning workbook",
            (
                "Detecting sheets, candidate force/displacement columns, units, labels, "
                "and geometry defaults. This does not calculate final metrics."
            ),
        )
        SCAN = fp.scan_workbook(CONFIG)

    CONFIG.sheet_mapping_file = SCAN["sheet_mapping_file"]
    SCAN_MAP = pd.read_csv(SCAN["sheet_mapping_file"])

    if "sheet" in SCAN_MAP.columns:
        SCAN_MAP["sheet"] = SCAN_MAP["sheet"].astype(str)

    if "notes" not in SCAN_MAP.columns:
        SCAN_MAP["notes"] = ""

    if "include" in SCAN_MAP.columns:
        SCAN_MAP["include"] = SCAN_MAP["include"].apply(_as_include_bool).astype(bool)

    for _unit_col in ["force_unit", "displacement_unit"]:
        if _unit_col in SCAN_MAP.columns:
            SCAN_MAP[_unit_col] = SCAN_MAP[_unit_col].apply(_clean_unit)

    try:
        TRACE_INPUTS_FOR_AUDIT = fp.build_trace_inputs(CONFIG, SCAN_MAP)
        TRACE_INTEGRITY_AUDIT = fp.trace_integrity_audit_from_traces(
            TRACE_INPUTS_FOR_AUDIT,
            CONFIG,
        )
    except Exception as exc:
        TRACE_INTEGRITY_AUDIT = pd.DataFrame()
        display(
            card(
                "Trace-integrity preflight could not complete",
                f"{type(exc).__name__}: {exc}",
                THEME["orange"],
                THEME["amber_bg"],
            )
        )

    if WIDGETS_AVAILABLE:
        final_status = widgets.Output()
        full_ui = widgets.VBox([])
        out = widgets.Output()
        review_out = widgets.Output()
        preview_out = widgets.Output()
        hist_out = widgets.Output()
        selection_status = widgets.Output()
        specific_issue_alert = widgets.HTML("")

        SCAN_CORRECTION_HISTORY = globals().setdefault("SCAN_CORRECTION_HISTORY", [])

        proceed_with_issues_ack = widgets.Checkbox(
            value=False,
            description="I reviewed unresolved mapping/trace-integrity items and want to proceed anyway.",
            layout=widgets.Layout(width="95%"),
        )

        def save_map():
            SCAN_MAP.to_csv(SCAN["sheet_mapping_file"], index=False)
            CONFIG.sheet_mapping_file = SCAN["sheet_mapping_file"]

        search_txt = widgets.Text(
            value="",
            placeholder="Search sheets, group, adhesive...",
            description="Search:",
            layout=widgets.Layout(width="95%"),
        )

        filter_dd = widgets.Dropdown(
            description="Filter:",
            layout=widgets.Layout(width="95%"),
        )

        select_all_btn = widgets.Button(
            description="Select all filtered",
            layout=widgets.Layout(width="95%"),
        )

        def _filter_options_and_default():
            masks = mapping_issue_masks(SCAN_MAP, TRACE_INTEGRITY_AUDIT)
            labels = [
                (f"All sheets ({int(masks['all'].sum())})", "all"),
                (f"Needs review ({int(masks['needs_review'].sum())})", "needs_review"),
                (f"Included sheets ({int(masks['included'].sum())})", "included"),
                (f"Excluded sheets ({int(masks['excluded'].sum())})", "excluded"),
                (f"Low-confidence columns ({int(masks['low_conf'].sum())})", "low_conf"),
                (
                    f"Trace-integrity warnings/failures ({int(masks['trace_issue'].sum())})",
                    "trace_issue",
                ),
                (
                    f"Missing labels/columns, invalid geometry, or invalid units ({int(masks['unknown'].sum())})",
                    "unknown",
                ),
                (
                    f"Invalid geometry ({int(masks['invalid_geometry'].sum())})",
                    "invalid_geometry",
                ),
                (
                    f"Invalid/missing units ({int(masks['invalid_units'].sum())})",
                    "invalid_units",
                ),
                (
                    f"Identical force/displacement columns ({int(masks['identical_cols'].sum())})",
                    "identical_cols",
                ),
            ]
            default = "needs_review" if int(masks["needs_review"].sum()) else "all"
            return labels, default

        def _update_filter_options(preserve=True):
            old = filter_dd.value if preserve and filter_dd.options else None
            labels, default = _filter_options_and_default()

            masks_now = mapping_issue_masks(SCAN_MAP, TRACE_INTEGRITY_AUDIT)
            if old == "needs_review" and int(masks_now["needs_review"].sum()) == 0:
                old = "all"

            filter_dd.options = labels
            allowed = [v for _, v in labels]
            filter_dd.value = old if old in allowed else default

        def _filtered_map():
            masks = mapping_issue_masks(SCAN_MAP, TRACE_INTEGRITY_AUDIT)
            key = filter_dd.value or "all"
            mask = masks.get(key, masks["all"])

            work = SCAN_MAP.loc[mask].copy()

            q = search_txt.value.strip().lower()
            if q:
                cols = [
                    c for c in work.columns
                    if c in ["sheet", "main_group", "adhesive", "notes", "warnings", "problem"]
                ]
                searchable = (
                    work[cols]
                    .astype(str)
                    .agg(" ".join, axis=1)
                    .str.lower()
                )
                work = work.loc[searchable.str.contains(q, regex=False)]

            return work

        def render_review_summary(*_):
            with review_out:
                clear_output(wait=True)

                try:
                    masks_now = mapping_issue_masks(SCAN_MAP, TRACE_INTEGRITY_AUDIT)
                    if (
                        filter_dd.value == "needs_review"
                        and int(masks_now["needs_review"].sum()) == 0
                    ):
                        filter_dd.value = "all"
                except Exception:
                    pass

                display_active_state()

                display(
                    card(
                        "Step 3 â€” Smart Mapping Review",
                        (
                            f"Mapping file: <code>{SCAN['sheet_mapping_file']}</code><br>"
                            "Use the split-panel editor below to resolve issues sheet by sheet, "
                            "or batch-apply fixes."
                        ),
                        THEME["blue"],
                        THEME["pale_blue"],
                    )
                )

                display(mapping_precheck_banner(SCAN_MAP, TRACE_INTEGRITY_AUDIT))

                fdf = _filtered_map()
                title = f"Current sheet mapping preview â€” {len(fdf)} row(s) shown"

                display_wrapped_table(
                    fdf[_display_cols(fdf)] if not fdf.empty else fdf,
                    title,
                    max_rows=10,
                )

                if isinstance(TRACE_INTEGRITY_AUDIT, pd.DataFrame) and not TRACE_INTEGRITY_AUDIT.empty:
                    tifail = int(
                        (
                            TRACE_INTEGRITY_AUDIT.get(
                                "trace_integrity_status",
                                pd.Series(dtype=str),
                            )
                            == "FAIL"
                        ).sum()
                    )
                    tiwarn = int(
                        (
                            TRACE_INTEGRITY_AUDIT.get(
                                "trace_integrity_status",
                                pd.Series(dtype=str),
                            )
                            == "WARN"
                        ).sum()
                    )

                    if tifail:
                        display(
                            card(
                                "Trace-integrity failures detected",
                                (
                                    f"{tifail} sheet(s) have severe preflight issues. "
                                    "Exclude/correct them before analysis."
                                ),
                                THEME["red"],
                                THEME["red_bg"],
                            )
                        )
                    elif tiwarn:
                        display(
                            card(
                                "Trace-integrity warnings detected",
                                (
                                    f"{tiwarn} sheet(s) have warnings such as large displacement "
                                    "gaps or missing values."
                                ),
                                THEME["orange"],
                                THEME["amber_bg"],
                            )
                        )
                    else:
                        display(
                            card(
                                "Trace-integrity preflight passed",
                                (
                                    "No severe missing/discontinuous displacement issues were "
                                    "detected in included sheets."
                                ),
                                THEME["green"],
                                THEME["green_bg"],
                            )
                        )

                    show_detail_button(
                        "Show trace-integrity audit",
                        lambda: TRACE_INTEGRITY_AUDIT,
                        button_style="",
                    )

                    display(
                        help_widget(
                            "Trace-integrity audit",
                            (
                                "Shows missing/skipped/discontinuous displacement checks for each trace. "
                                "Use this when the mapping precheck reports trace-integrity warnings or failures."
                            ),
                            width="400px",
                        )
                    )

        def refresh_tables():
            with preview_out:
                clear_output(wait=True)

                if len(SCAN_CORRECTION_HISTORY) == 0:
                    display(
                        HTML(
                            "<div class='peel-small-note'>No manual corrections applied yet.</div>"
                        )
                    )
                else:
                    fdf = _filtered_map()
                    display_wrapped_table(
                        fdf[_display_cols(fdf)] if not fdf.empty else fdf,
                        "Updated mapping preview",
                        max_rows=10,
                    )

            with hist_out:
                clear_output(wait=True)

                if len(SCAN_CORRECTION_HISTORY) > 0:
                    display_wrapped_table(
                        pd.DataFrame(SCAN_CORRECTION_HISTORY),
                        "Correction history",
                        max_rows=40,
                    )

        all_sheets = list(SCAN_MAP["sheet"].astype(str))

        sheet_sel = widgets.SelectMultiple(
            options=all_sheets,
            value=tuple(all_sheets[:1]),
            description="",
            layout=widgets.Layout(width="95%", height="240px"),
        )

        def _visible_filtered_sheet_names():
            fdf = _filtered_map()
            allowed = set(all_sheets)

            if "sheet" not in fdf.columns:
                return []

            return [
                x for x in list(fdf["sheet"].astype(str))
                if x in allowed
            ]

        def select_all(*args):
            vals = tuple(_visible_filtered_sheet_names())
            if vals:
                sheet_sel.options = list(vals)
                sheet_sel.value = vals

        select_all_btn.on_click(select_all)

        include_mode = widgets.Dropdown(
            options=[
                ("leave unchanged", ""),
                ("Include selected sheets", "include"),
                ("Exclude selected sheets", "exclude"),
            ],
            description="Include:",
            layout=widgets.Layout(width="95%"),
        )

        group_txt = widgets.Text(
            description="Group:",
            layout=widgets.Layout(width="47%"),
        )

        adhesive_txt = widgets.Text(
            description="Adhesive:",
            layout=widgets.Layout(width="47%"),
        )

        rep_txt = widgets.Text(
            description="Replicate:",
            layout=widgets.Layout(width="47%"),
        )

        replicate_note = widgets.HTML("")

        width_ft = widgets.FloatText(
            value=WIDTH_MM,
            description="w mm:",
            layout=widgets.Layout(width="31%"),
        )

        thick_ft = widgets.FloatText(
            value=THICKNESS_MM,
            description="t mm:",
            layout=widgets.Layout(width="31%"),
        )

        bond_ft = widgets.FloatText(
            value=BOND_LENGTH_MM,
            description="bond mm:",
            layout=widgets.Layout(width="31%"),
        )

        notes_txt = widgets.Text(
            description="Notes:",
            layout=widgets.Layout(width="95%"),
        )

        force_dd = widgets.Dropdown(
            description="Force col:",
            layout=widgets.Layout(width="47%"),
        )

        disp_dd = widgets.Dropdown(
            description="Disp col:",
            layout=widgets.Layout(width="47%"),
        )

        force_unit = widgets.Dropdown(
            options=["leave unchanged", "N", "mN", "lbf"],
            description="Force unit:",
            layout=widgets.Layout(width="47%"),
        )

        disp_unit = widgets.Dropdown(
            options=["leave unchanged", "mm", "cm", "in", "um"],
            description="Disp unit:",
            layout=widgets.Layout(width="47%"),
        )

        t_peel_alert = widgets.Output()


        def get_sheet_issues(sheet_name):
            issues = []

            match = SCAN_MAP.loc[SCAN_MAP["sheet"].astype(str) == str(sheet_name)]

            if match.empty:
                return ["sheet not found in mapping table"]

            row = match.iloc[0]

            if not _as_include_bool(row.get("include", True)):
                return ["Sheet is excluded (No issues)"]

            # Column-confidence warning
            conf = str(row.get("column_confidence", "")).strip().upper()
            if conf and conf not in {"HIGH", "OK", "PASS"}:
                issues.append(f"low-confidence column assignment ({conf})")

            # Scanner warning/problem fields
            warnings_text = str(row.get("warnings", "")).strip()
            problem_text = str(row.get("problem", "")).strip()

            if warnings_text and warnings_text.lower() not in {"nan", "none"}:
                issues.append(f"scanner warning: {warnings_text}")

            if problem_text and problem_text.lower() not in {"nan", "none"}:
                issues.append(f"scanner problem: {problem_text}")

            # Geometry
            w = pd.to_numeric(row.get("width_mm"), errors="coerce")
            t = pd.to_numeric(row.get("thickness_mm"), errors="coerce")
            b = pd.to_numeric(row.get("bond_length_mm"), errors="coerce")

            if pd.isna(w):
                issues.append("missing width")
            elif float(w) <= 0:
                issues.append("width must be > 0")

            if pd.isna(t):
                issues.append("missing thickness")
            elif float(t) <= 0:
                issues.append("thickness must be > 0")

            if pd.isna(b):
                issues.append("missing bond length")
            elif float(b) <= 0:
                issues.append("bond length must be > 0")

            # Labels
            if _is_missing_like(row.get("main_group")):
                issues.append("missing group")

            if _is_missing_like(row.get("adhesive")):
                issues.append("missing adhesive")

            # Columns
            f_col = _clean_col(row.get("force_column"))
            d_col = _clean_col(row.get("displacement_column"))

            if not f_col:
                issues.append("missing force column")

            if not d_col:
                issues.append("missing displacement column")

            if f_col and d_col and f_col == d_col:
                issues.append("force/displacement columns are identical")

            # Units
            force_u = _clean_unit(row.get("force_unit"))
            disp_u = _clean_unit(row.get("displacement_unit"))

            if not force_u:
                issues.append("missing force unit")
            elif force_u not in _VALID_FORCE_UNITS:
                issues.append(f"unsupported force unit ({force_u})")

            if not disp_u:
                issues.append("missing displacement unit")
            elif disp_u not in _VALID_DISPLACEMENT_UNITS:
                issues.append(f"unsupported displacement unit ({disp_u})")

            # Trace integrity
            status_map = _trace_status_by_sheet(TRACE_INTEGRITY_AUDIT)
            t_status = status_map.get(str(sheet_name), "")

            if t_status == "FAIL":
                issues.append("TRACE INTEGRITY FAILED (consider exclude/correct)")
            elif t_status == "WARN":
                issues.append("trace warning; check audit table")

            return issues

        def load_from_first_selected(*args):
            selected = list(sheet_sel.value)

            if not selected:
                specific_issue_alert.value = ""
                return

            sheet = selected[0]

            if len(selected) > 1:
                specific_issue_alert.value = (
                    f"<div style='padding:8px; border-left:4px solid {THEME['blue']}; "
                    f"background:{THEME['pale_blue']}; margin-bottom:10px; border-radius:4px;'>"
                    f"<b>Batch Mode:</b> Editing {len(selected)} sheets. "
                    "Applied values will overwrite all selected sheets.</div>"
                )
            else:
                issues = get_sheet_issues(sheet)

                if not issues:
                    specific_issue_alert.value = (
                        f"<div style='padding:8px; border-left:4px solid {THEME['green']}; "
                        f"background:{THEME['green_bg']}; margin-bottom:10px; border-radius:4px;'>"
                        f"<b>Ready:</b> {sheet} has no detected mapping issues.</div>"
                    )
                elif "Sheet is excluded" in issues[0]:
                    specific_issue_alert.value = (
                        f"<div style='padding:8px; border-left:4px solid {THEME['gray']}; "
                        f"background:{THEME['gray_bg']}; margin-bottom:10px; border-radius:4px;'>"
                        "<b>Excluded:</b> This sheet will be skipped during analysis.</div>"
                    )
                else:
                    color = THEME["red"] if "FAILED" in str(issues) else THEME["orange"]
                    bg = THEME["red_bg"] if "FAILED" in str(issues) else THEME["amber_bg"]

                    specific_issue_alert.value = (
                        f"<div style='padding:8px; border-left:4px solid {color}; "
                        f"background:{bg}; margin-bottom:10px; border-radius:4px;'>"
                        f"<b>Action Required for {sheet}:</b> "
                        f"missing/check {', '.join(issues)}</div>"
                    )

            row = SCAN_MAP.loc[SCAN_MAP["sheet"].astype(str) == str(sheet)].iloc[0]

            group_txt.value = "" if _is_missing_like(row.get("main_group", "")) else str(row.get("main_group", ""))
            adhesive_txt.value = "" if _is_missing_like(row.get("adhesive", "")) else str(row.get("adhesive", ""))
            rep_txt.value = "" if _is_missing_like(row.get("replicate_id", "")) else str(row.get("replicate_id", ""))
            notes_txt.value = "" if _is_missing_like(row.get("notes", "")) else str(row.get("notes", ""))

            width_ft.value = _safe_positive_float(row.get("width_mm"), WIDTH_MM)
            thick_ft.value = _safe_positive_float(row.get("thickness_mm"), THICKNESS_MM)
            bond_ft.value = _safe_positive_float(row.get("bond_length_mm"), BOND_LENGTH_MM)

            try:
                df = fp.read_full_table_for_scan(INPUT_PATH, sheet)
                cols = list(df.columns)
            except Exception:
                cols = [
                    str(row.get("force_column", "")),
                    str(row.get("displacement_column", "")),
                ]

            force_dd.options = ["leave unchanged"] + cols
            disp_dd.options = ["leave unchanged"] + cols

            current_force = row.get("force_column")
            current_disp = row.get("displacement_column")

            force_dd.value = (
                current_force
                if current_force in cols and len(selected) == 1
                else "leave unchanged"
            )

            disp_dd.value = (
                current_disp
                if current_disp in cols and len(selected) == 1
                else "leave unchanged"
            )

            force_unit.value = "leave unchanged"
            disp_unit.value = "leave unchanged"
            include_mode.value = ""

            if len(selected) > 1:
                rep_txt.layout.display = "none"
                replicate_note.value = (
                    "<span class='peel-small-note' style='margin-left:10px;'>"
                    "Replicate ID is hidden for multi-sheet selection.</span>"
                )
            else:
                rep_txt.layout.display = "inline-flex"
                replicate_note.value = ""

            with t_peel_alert:
                clear_output(wait=False)

                if "t-peel" in adhesive_txt.value.lower() or "t-peel" in notes_txt.value.lower():
                    display(
                        card(
                            "T-peel geometry check",
                            (
                                "T-peel displacement can differ from bonded-overlap length. "
                                "Confirm <b>bond mm</b> is the intended bonded overlap."
                            ),
                            THEME["orange"],
                            THEME["amber_bg"],
                        )
                    )

        sheet_sel.observe(load_from_first_selected, names="value")

        def select_filtered_sheets(*_):
            try:
                names = _visible_filtered_sheet_names()

                if names:
                    old_visible = tuple([x for x in sheet_sel.value if x in names])
                    sheet_sel.options = names

                    if old_visible:
                        sheet_sel.value = old_visible[:1]
                    else:
                        sheet_sel.value = (names[0],)
                else:
                    sheet_sel.options = []
                    sheet_sel.value = tuple()

            except Exception:
                pass

            render_review_summary()
            refresh_tables()

        search_txt.observe(select_filtered_sheets, names="value")
        filter_dd.observe(select_filtered_sheets, names="value")

        _update_filter_options(preserve=False)
        select_filtered_sheets()
        load_from_first_selected()

        def record_change(sheet, field, old, new):
            if str(old) != str(new):
                SCAN_CORRECTION_HISTORY.append(
                    {
                        "correction_no": len(SCAN_CORRECTION_HISTORY) + 1,
                        "sheet": sheet,
                        "field": field,
                        "old_value": old,
                        "new_value": new,
                    }
                )

        apply_btn = widgets.Button(
            description="Apply changes",
            button_style="warning",
            layout=widgets.Layout(width="180px"),
        )

        exclude_btn = widgets.Button(
            description="Exclude selected",
            button_style="danger",
            layout=widgets.Layout(width="180px"),
        )

        confirm_btn = widgets.Button(
            description="Confirm mapping",
            button_style="primary",
            layout=widgets.Layout(width="280px"),
        )

        def _set_action_buttons(disabled):
            for b in [apply_btn, exclude_btn, confirm_btn]:
                b.disabled = bool(disabled)

        def _editor_validation_message(selected):
            for label, value in [
                ("width", width_ft.value),
                ("thickness", thick_ft.value),
                ("bond length", bond_ft.value),
            ]:
                x = pd.to_numeric(value, errors="coerce")

                if pd.isna(x) or float(x) <= 0:
                    return f"{label} must be greater than 0."

            for sheet in selected:
                match = SCAN_MAP.loc[SCAN_MAP["sheet"].astype(str) == str(sheet)]

                if match.empty:
                    continue

                row = match.iloc[0]

                final_force = (
                    force_dd.value
                    if force_dd.value != "leave unchanged"
                    else row.get("force_column")
                )

                final_disp = (
                    disp_dd.value
                    if disp_dd.value != "leave unchanged"
                    else row.get("displacement_column")
                )

                f_col = _clean_col(final_force)
                d_col = _clean_col(final_disp)

                if f_col and d_col and f_col == d_col:
                    return f"{sheet}: force and displacement columns cannot be identical."

            return ""

        def _apply_common(selected, force_exclude=False):
            if not selected:
                status_in(
                    out,
                    card(
                        "No sheets selected",
                        "Select one or more sheets first.",
                        THEME["orange"],
                        THEME["amber_bg"],
                    ),
                )
                return

            if not force_exclude:
                validation_msg = _editor_validation_message(selected)

                if validation_msg:
                    status_in(
                        out,
                        card(
                            "Cannot apply mapping update",
                            validation_msg,
                            THEME["red"],
                            THEME["red_bg"],
                        ),
                    )
                    return

            was_needs_review = filter_dd.value == "needs_review"
            first_sheet = selected[0]

            _set_action_buttons(True)

            try:
                status_in(
                    out,
                    card(
                        "Applying mapping update",
                        f"Updating {len(selected)} sheet(s).",
                        "#546e7a",
                        "#f8fafb",
                    ),
                )

                for sheet in selected:
                    match_idx = SCAN_MAP.index[SCAN_MAP["sheet"].astype(str) == str(sheet)]

                    if len(match_idx) == 0:
                        continue

                    idx = match_idx[0]

                    vals = {
                        "width_mm": float(width_ft.value),
                        "thickness_mm": float(thick_ft.value),
                        "bond_length_mm": float(bond_ft.value),
                    }

                    if force_exclude:
                        vals["include"] = False
                    elif include_mode.value == "include":
                        vals["include"] = True
                    elif include_mode.value == "exclude":
                        vals["include"] = False

                    if group_txt.value.strip():
                        vals["main_group"] = group_txt.value.strip()

                    if adhesive_txt.value.strip():
                        vals["adhesive"] = adhesive_txt.value.strip()

                    if rep_txt.value.strip() and len(selected) == 1:
                        vals["replicate_id"] = rep_txt.value.strip()

                    if notes_txt.value.strip():
                        vals["notes"] = notes_txt.value.strip()

                    if force_unit.value != "leave unchanged":
                        vals["force_unit"] = force_unit.value

                    if disp_unit.value != "leave unchanged":
                        vals["displacement_unit"] = disp_unit.value

                    if force_dd.value != "leave unchanged":
                        vals["force_column"] = force_dd.value

                    if disp_dd.value != "leave unchanged":
                        vals["displacement_column"] = disp_dd.value

                    for k, v in vals.items():
                        if k not in SCAN_MAP.columns:
                            SCAN_MAP[k] = ""

                        old = SCAN_MAP.loc[idx, k]
                        record_change(sheet, k, old, v)
                        SCAN_MAP.loc[idx, k] = v

                save_map()

                try:
                    trace_inputs = fp.build_trace_inputs(CONFIG, SCAN_MAP)
                    globals()["TRACE_INTEGRITY_AUDIT"] = fp.trace_integrity_audit_from_traces(
                        trace_inputs,
                        CONFIG,
                    )
                except Exception:
                    pass

                _update_filter_options(preserve=True)

                if was_needs_review:
                    masks_now = mapping_issue_masks(SCAN_MAP, TRACE_INTEGRITY_AUDIT)
                    remaining = SCAN_MAP.loc[masks_now["needs_review"]].copy()

                    if not remaining.empty:
                        current_row = SCAN_MAP.loc[
                            SCAN_MAP["sheet"].astype(str) == str(first_sheet)
                        ].iloc[0]

                        current_group = str(current_row.get("main_group", "")).strip()
                        current_adhesive = str(current_row.get("adhesive", "")).strip()

                        same_group = pd.DataFrame()
                        if current_group and "main_group" in remaining.columns:
                            same_group = remaining[
                                remaining["main_group"].astype(str).str.strip() == current_group
                            ]

                        same_adhesive = pd.DataFrame()
                        if current_adhesive and "adhesive" in remaining.columns:
                            same_adhesive = remaining[
                                remaining["adhesive"].astype(str).str.strip() == current_adhesive
                            ]

                        if not same_group.empty:
                            next_sheet = str(same_group["sheet"].iloc[0])
                        elif not same_adhesive.empty:
                            next_sheet = str(same_adhesive["sheet"].iloc[0])
                        else:
                            next_sheet = str(remaining["sheet"].iloc[0])

                        if filter_dd.value != "needs_review":
                            filter_dd.value = "needs_review"

                        if next_sheet not in list(sheet_sel.options):
                            sheet_sel.options = list(remaining["sheet"].astype(str))

                        sheet_sel.value = (next_sheet,)

                        status_in(
                            out,
                            card(
                                "Success: Auto-Advanced",
                                (
                                    f"Saved {first_sheet}. Automatically loaded the next sheet "
                                    f"requiring review: <b>{next_sheet}</b>."
                                ),
                                THEME["green"],
                                THEME["green_bg"],
                            ),
                        )

                    else:
                        filter_dd.value = "all"
                        sheet_sel.options = all_sheets
                        sheet_sel.value = tuple(all_sheets[:1]) if all_sheets else tuple()

                        status_in(
                            out,
                            card(
                                "All Review Items Resolved",
                                (
                                    "The filter has switched back to 'All sheets'. "
                                    "You may now confirm the mapping."
                                ),
                                THEME["green"],
                                THEME["green_bg"],
                            ),
                        )

                else:
                    status_in(
                        out,
                        card(
                            "Mapping update applied",
                            f"Updated {len(selected)} sheet(s).",
                            THEME["green"],
                            THEME["green_bg"],
                        ),
                    )

                render_review_summary()
                refresh_tables()
                load_from_first_selected()

            finally:
                _set_action_buttons(False)

        apply_btn.on_click(lambda _: _apply_common(list(sheet_sel.value), False))
        exclude_btn.on_click(lambda _: _apply_common(list(sheet_sel.value), True))

        def unresolved_mapping_review_count():
            try:
                return int(mapping_issue_masks(SCAN_MAP, TRACE_INTEGRITY_AUDIT)["needs_review"].sum())
            except Exception:
                return 0

        def on_confirm(_):
            global SCAN_CONFIRMED, MAPPING_CHANGED, RESULTS_STALE

            unresolved = unresolved_mapping_review_count()

            if unresolved > 0 and not proceed_with_issues_ack.value:
                status_in(
                    final_status,
                    final_banner(
                        "Mapping review still needed",
                        (
                            f"{unresolved} included sheet(s) still have mapping or "
                            "trace-integrity review items. Resolve them, exclude affected sheets, "
                            "or check the acknowledgment box to proceed anyway."
                        ),
                        "warn",
                    ),
                )
                return

            save_map()

            SCAN_CONFIRMED = True
            MAPPING_CHANGED = True
            RESULTS_STALE = True

            try:
                full_ui.layout.display = "none"
            except Exception:
                pass

            for area in [review_out, preview_out, hist_out, out, selection_status]:
                with area:
                    clear_output(wait=False)

            if "include" in SCAN_MAP.columns:
                n_inc = int(SCAN_MAP["include"].apply(_as_include_bool).sum())
            else:
                n_inc = len(SCAN_MAP)

            n_exc = len(SCAN_MAP) - n_inc

            status_in(
                final_status,
                final_banner(
                    "Mapping confirmed",
                    (
                        f"{n_inc} sheet(s) included, {n_exc} excluded.<br>"
                        "Next: run Step 4 to activate an analysis profile.<br><br>"
                        "<b>To change mapping, rerun Step 3.</b>"
                    ),
                    "ok",
                ),
            )

        confirm_btn.on_click(on_confirm)

        left_panel = widgets.VBox(
            [
                widgets.HTML("<b>Filter & Select</b>"),
                search_txt,
                filter_dd,
                select_all_btn,
                widgets.HTML(
                    "<span style='font-size:11px;color:#777;margin-top:6px;display:block;'>"
                    "(Ctrl/Cmd + click for multi-select)</span>"
                ),
                sheet_sel,
            ],
            layout=widgets.Layout(width="35%"),
        )

        right_panel = widgets.VBox(
            [
                specific_issue_alert,
                include_mode,
                widgets.HBox(
                    [group_txt, adhesive_txt],
                    layout=widgets.Layout(justify_content="space-between", width="95%"),
                ),
                widgets.HBox(
                    [width_ft, thick_ft, bond_ft],
                    layout=widgets.Layout(justify_content="space-between", width="95%"),
                ),
                widgets.HBox(
                    [force_dd, disp_dd],
                    layout=widgets.Layout(justify_content="space-between", width="95%"),
                ),
                widgets.HBox(
                    [force_unit, disp_unit],
                    layout=widgets.Layout(justify_content="space-between", width="95%"),
                ),
                widgets.HBox(
                    [rep_txt, replicate_note],
                    layout=widgets.Layout(width="95%"),
                ),
                notes_txt,
                t_peel_alert,
                widgets.HTML(
                    "<div style='margin-top:10px; border-top:1px solid #eee; padding-top:10px;'></div>"
                ),
                proceed_with_issues_ack,
                widgets.HBox(
                    [
                        apply_btn,
                        help_widget(
                            "Apply changes",
                            (
                                "Saves the visible field values to the selected sheet(s). "
                                "If you are in the 'Needs review' filter, clicking this will "
                                "automatically advance to the next broken sheet."
                            ),
                            width="250px",
                        ),
                        exclude_btn,
                        help_widget(
                            "Exclude selected",
                            (
                                "Marks selected sheets as excluded without changing the raw workbook. "
                                "Excluded sheets bypass trace-integrity requirements."
                            ),
                            width="250px",
                        ),
                    ]
                ),
                widgets.HBox(
                    [
                        confirm_btn,
                        help_widget(
                            "Confirm mapping",
                            (
                                "Locks the current mapping and collapses this entire panel so you can "
                                "proceed to Step 4. You cannot confirm if there are unresolved issues "
                                "unless you check the acknowledge box."
                            ),
                            width="250px",
                        ),
                    ]
                ),
            ],
            layout=widgets.Layout(
                width="65%",
                padding="0 0 0 15px",
                border_left="1px solid #ddd",
            ),
        )

        split_editor = widgets.HBox(
            [left_panel, right_panel],
            layout=widgets.Layout(
                margin="10px 0",
                border="1px solid #cfd8dc",
                padding="15px",
                border_radius="8px",
            ),
        )

        full_ui.children = [
            review_out,
            split_editor,
            out,
            preview_out,
            hist_out,
        ]

        render_review_summary()
        refresh_tables()

        status_in(
            final_status,
            final_banner(
                "Waiting for mapping confirmation",
                (
                    "Review the mapping table and precheck banner. Resolve issues using the "
                    "editor above, then confirm."
                ),
                "warn",
            ),
        )

        display(final_status)
        display(full_ui)

    else:
        CONFIG.sheet_mapping_file = SCAN["sheet_mapping_file"]
        SCAN_CONFIRMED = True
        MAPPING_CHANGED = True
        RESULTS_STALE = True

        ok_card(
            "Mapping confirmed",
            "Widgets unavailable; detected mapping was accepted directly. Run Step 4.",
        )

step_timer_end("Step 3 mapping and trace-integrity review", _t0)
```

```python
_t0 = step_timer_start("Step 4 profile selection")
require_reference_validation()
#@title Step 4 â€” Choose analysis profile
#@markdown Select the read-only manuscript baseline or explicitly load/create a NON_MANUSCRIPT_MODIFIED user profile. This step uses the finalized Step 3 mapping to warn about window feasibility before activation.

# Re-opening Step 4
globals()["SETTINGS_CONFIRMED"] = False

if not globals().get("SCAN_CONFIRMED", False):
    waiting("Step 4 waiting for finalized mapping", "Return to Step 3 and click <b>Confirm all mapping decisions</b>.")
else:
    def make_baseline_config():
        cfg = fp.build_manuscript_baseline_config(
            input_path=INPUT_PATH,
            output_root="Peel_Analysis_Output_v1_4_0_rc15",
            width_mm=WIDTH_MM,
            thickness_mm=THICKNESS_MM,
            bond_length_mm=BOND_LENGTH_MM,
        )
        cfg.sheet_mapping_file = SCAN["sheet_mapping_file"]
        return cfg

    def current_mapping_df():
        return pd.read_csv(SCAN["sheet_mapping_file"])

    def _window_feasibility_tables(window_mm=25.0, start_mm=5.0, tail_frac=0.20):
        cfg = make_baseline_config()
        cfg.window_length_mm = float(window_mm)
        cfg.start_offset_mm = float(start_mm)
        cfg.tail_exclude_frac = float(tail_frac)
        mapping = current_mapping_df()
        summary = fp.feasible_window_summary_from_mapping(mapping, cfg)
        affected = fp.infeasible_sheets_from_mapping(mapping, cfg)
        if isinstance(affected, pd.DataFrame) and not affected.empty and "status" in affected.columns:
            affected = affected[affected["status"].isin(["WINDOW_TOO_LONG_FOR_SHEET", "ADVANCED_SHORT_WINDOW"])]
        return summary, affected

    def _render_window_feasibility(window_mm, start_mm, tail_frac, out=None):
        target = out
        try:
            summary, affected = _window_feasibility_tables(window_mm, start_mm, tail_frac)
            def _show():
                display_wrapped_table(summary, "Window feasibility from current mapping", max_rows=10)
                active = float(window_mm)
                floor = 12.5
                if isinstance(affected, pd.DataFrame) and not affected.empty:
                    display(card("Window-length caution", f"Selected window = <b>{active:.2f} mm</b>. One or more included sheets may not support this window after start-offset and tail-exclusion guards. Shorten the window, exclude affected sheets in Step 3, or use manuscript_baseline_v1 only if paper-replica analysis is required.", THEME["orange"], THEME["amber_bg"]))
                    cols = [c for c in ["sheet","rough_displacement_span_mm","estimated_feasible_window_mm","active_window_length_mm","status","message"] if c in affected.columns]
                    display_wrapped_table(affected[cols], "Included sheets affected by the current window choice", max_rows=15)
                elif active < floor:
                    display(card("Advanced short-window caution", f"Selected window = <b>{active:.2f} mm</b>, below half of the manuscript-baseline 25 mm window ({floor:.1f} mm). Treat this as advanced/manual sensitivity analysis, not manuscript-replica output.", THEME["orange"], THEME["amber_bg"]))
                else:
                    display(card("Window feasibility check", "The selected window appears feasible for the currently included mapped sheets. Continue only after confirming the profile purpose and labels.", THEME["green"], THEME["green_bg"]))

            if target is None: _show()
            else:
                with target:
                    clear_output(wait=True)
                    _show()
            return summary, affected
        except Exception as exc:
            msg = card("Window feasibility unavailable", f"<code>{type(exc).__name__}: {exc}</code>", THEME["orange"], THEME["amber_bg"])
            if target is None: display(msg)
            else:
                with target:
                    clear_output(wait=True)
                    display(msg)
            return pd.DataFrame(), pd.DataFrame()

    info_card(
        "Step 4 â€” Active method profile",
        "Choose one profile for the next analysis. <b>manuscript_baseline_v1</b> is read-only and reproduces the paper-guided settings. User profiles are separate, editable, and always labeled <b>NON_MANUSCRIPT_MODIFIED</b>."
    )
    display_active_state()
    baseline_summary = pd.DataFrame([{
        "profile":"manuscript_baseline_v1", "type":"built-in read-only", "window_mm":25.0, "start_mm":5.0, "tail_frac":0.20,
        "drift_limit":0.25, "prom_frac":"0.003 Ã— Fmax", "strict_extrema_spread":"advisory/off", "note":"Use for manuscript-replica outputs."
    }])
    display_wrapped_table(baseline_summary, "Available built-in profile", max_rows=5)

    if WIDGETS_AVAILABLE:
        out = widgets.Output()

        # --- UI CONTAINERS ---
        mode = widgets.ToggleButtons(
            options=[("Use manuscript baseline", "baseline"), ("Load user profile", "load"), ("Create/edit profile", "edit")],
            value="baseline", layout=widgets.Layout(width="760px"), style={"button_width": "240px"}
        )
        mode_bar = widgets.HBox([mode, help_widget("Need help choosing a profile?", "Use manuscript_baseline_v1 for paper-replica outputs. Load a saved user profile only when you have a documented NON_MANUSCRIPT_MODIFIED profile from a previous session. Create/edit a profile only for sensitivity, recovery, or a different dataset type.", width="430px")])

        dynamic_panel = widgets.VBox([]) # This instantly swaps the UI panels
        main_ui_box = widgets.VBox([mode_bar, dynamic_panel])

        # --- BASELINE UI ---
        btn_base = widgets.Button(description="Use manuscript baseline", button_style="primary", layout=widgets.Layout(width="300px"))
        ui_base = widgets.VBox([
            widgets.HTML(card("manuscript_baseline_v1", "Built-in read-only baseline. The edit panel is hidden. Click the button below to activate it for the next analysis.", THEME["blue"], THEME["pale_blue"]).data),
            widgets.HBox([btn_base, help_widget("Use manuscript baseline", "Activates locked manuscript_baseline_v1 settings. Use this for paper-replica outputs and manuscript-consistent analysis.", width="340px")])
        ])

        # --- LOAD UI ---
        profile_files = profile_file_candidates(".")
        profile_dd = widgets.Dropdown(options=profile_files or [""], description="Profile:", layout=widgets.Layout(width="95%"))
        profile_path_txt = widgets.Text(value="", placeholder="Path to JSON", description="Path:", layout=widgets.Layout(width="78%"))
        add_profile_btn = widgets.Button(description="Add/preview profile path", layout=widgets.Layout(width="230px"))
        profile_upload = widgets.FileUpload(accept=".json", multiple=False, description="Upload JSON")

        profile_preview_out = widgets.Output()
        profile_msg = widgets.Output()
        btn_load = widgets.Button(description="Activate loaded profile", button_style="primary", layout=widgets.Layout(width="300px"))

        load_header = widgets.HTML(card("Saved user profiles detected", f"{len(profile_files)} profile file(s) found. Select one to preview before activation.", THEME["blue"], THEME["pale_blue"]).data) if profile_files else widgets.HTML(card("No saved user profiles detected", "Use manuscript_baseline_v1, upload/place a <code>user_profile_*.json</code> file, or create a new user profile.", THEME["orange"], THEME["amber_bg"]).data)

        ui_load = widgets.VBox([
            load_header,
            profile_dd,
            profile_preview_out,
            widgets.HTML("<br><b>Optional: add a saved user-profile JSON</b>"),
            widgets.HBox([profile_path_txt, add_profile_btn]),
            profile_upload,
            profile_msg,
            widgets.HBox([btn_load, help_widget("Activate loaded profile", "Activates the previewed JSON profile for the next run. Outputs are labeled NON_MANUSCRIPT_MODIFIED.", width="340px")])
        ])

        # --- EDIT UI ---
        profile_id_w = widgets.Text(value="custom_profile", description="Short ID:", layout=widgets.Layout(width="360px"))
        profile_display_w = widgets.Text(value="User non-manuscript profile", description="Display:", layout=widgets.Layout(width="520px"))
        profile_note_w = widgets.Textarea(value="", description="Note:", layout=widgets.Layout(width="95%", height="70px"))
        win_w = widgets.FloatText(value=25.0, description="window mm:", layout=widgets.Layout(width="240px"))
        start_w = widgets.FloatText(value=5.0, description="start mm:", layout=widgets.Layout(width="240px"))
        tail_w = widgets.FloatText(value=0.20, description="tail frac:", layout=widgets.Layout(width="240px"))
        drift_w = widgets.FloatText(value=0.25, description="drift limit:", layout=widgets.Layout(width="240px"))
        prom_w = widgets.FloatText(value=0.003, description="prom frac:", layout=widgets.Layout(width="240px"))
        spread_w = widgets.Checkbox(value=False, description="strict extrema-spread QC")

        ack_w = widgets.Checkbox(value=False, description="I understand this creates NON_MANUSCRIPT_MODIFIED outputs that should not be mixed with manuscript-baseline results.", layout=widgets.Layout(width="95%"))
        short_window_ack_w = widgets.Checkbox(value=False, description="If applicable, I understand that a window below 12.5 mm is advanced/manual sensitivity analysis.", layout=widgets.Layout(width="95%"))
        infeasible_window_ack_w = widgets.Checkbox(value=False, description="If applicable, I understand that this recovery profile may fail mapped sheets whose feasible travel is shorter than the selected window.", layout=widgets.Layout(width="95%"))

        feas_out = widgets.Output()
        profile_guide_out = widgets.Output()
        btn_edit = widgets.Button(description="Save and activate user profile", button_style="primary", layout=widgets.Layout(width="300px"))
        guide_btn = widgets.Button(description="Show profile-setting guide", layout=widgets.Layout(width="250px"))

        profile_guide_df = pd.DataFrame([
            {"setting":"window_length_mm", "manuscript_baseline":"25.0 mm", "what_it_controls":"Length of the selected force-history window", "if_too_large":"Short traces may fail", "if_too_small":"Metrics become more local/sensitive"},
            {"setting":"start_offset_mm", "manuscript_baseline":"5.0 mm", "what_it_controls":"Initial displacement excluded before window search", "if_too_large":"May skip valid peel data", "if_too_small":"May include toe/slack behavior"},
            {"setting":"tail_exclude_frac", "manuscript_baseline":"0.20", "what_it_controls":"Final trace fraction excluded from stable-window search", "if_too_large":"Less usable travel", "if_too_small":"Terminal failure may be selected"},
            {"setting":"max_window_drift_frac", "manuscript_baseline":"0.25", "what_it_controls":"Normalized drift/ramp gate", "if_too_large":"Ramping windows may pass", "if_too_small":"Valid but imperfect windows may fail"},
            {"setting":"peak_prominence_frac", "manuscript_baseline":"0.003", "what_it_controls":"Peak/trough sensitivity", "if_too_large":"Real extrema may be missed", "if_too_small":"Noise may be counted"},
            {"setting":"strict_extrema_spread_check", "manuscript_baseline":"advisory/off", "what_it_controls":"Clustered-extrema treatment", "if_too_large":"N/A", "if_too_small":"N/A"},
        ])

        def _show_guide(_):
            with profile_guide_out:
                clear_output(wait=False)
                display_wrapped_table(profile_guide_df, "Profile setting guide", max_rows=10)
                close_guide = widgets.Button(description="Close guide", layout=widgets.Layout(width="130px"))
                close_guide.on_click(lambda __: profile_guide_out.clear_output())
                display(close_guide)
        guide_btn.on_click(_show_guide)

        ui_edit = widgets.VBox([
            widgets.HTML(card("Create/edit user profile", "Change only parameters needed for your dataset. Enter only a short ID; the file will be saved as <code>user_profile_&lt;profile_id&gt;.json</code>.", THEME["orange"], THEME["amber_bg"]).data),
            widgets.HBox([guide_btn, help_widget("Profile-setting guide", "Explains what each editable profile setting controls.", width="340px")]),
            profile_guide_out,
            widgets.HBox([profile_id_w, help_widget("Short profile ID", "Type only a short ID such as short_trace_jb.", width="360px")]),
            profile_display_w, profile_note_w,
            widgets.HBox([win_w, help_widget("Window length", "Length of the displacement region used for stable-window selection.", width="330px"), start_w, help_widget("Start offset", "Initial displacement excluded before window search.", width="330px")]),
            widgets.HBox([tail_w, help_widget("Tail exclusion", "Fraction of terminal displacement excluded from window search.", width="330px"), drift_w, help_widget("Drift limit", "Maximum allowed |slope|.", width="330px")]),
            widgets.HBox([prom_w, help_widget("Peak prominence", "Peak/trough threshold.", width="330px"), spread_w, help_widget("Strict extrema-spread QC", "Clustered extrema treatment.", width="330px")]),
            feas_out,
            widgets.HTML(card("Final user-profile activation", "Review the feasibility table and any affected-sheet warnings. Required acknowledgments appear directly below.", THEME["orange"], THEME["amber_bg"]).data),
            ack_w, short_window_ack_w, infeasible_window_ack_w,
            widgets.HBox([btn_edit, help_widget("Save and activate user profile", "Validates the profile, saves it, and activates it.", width="430px")])
        ])

        # --- LOGIC & OBSERVERS ---
        def profile_preview(path):
            try:
                data = fp.load_user_method_profile(path)
                keys = ["profile_id", "display_name", "profile_note", "window_length_mm", "start_offset_mm", "tail_exclude_frac", "max_window_drift_frac", "peak_prominence_frac", "strict_extrema_spread_check"]
                return pd.DataFrame([{k: data.get(k, "") for k in keys}])
            except Exception as exc: return pd.DataFrame([{"error": f"{type(exc).__name__}: {exc}"}])

        def refresh_feasibility(*_):
            return _render_window_feasibility(win_w.value, start_w.value, tail_w.value, feas_out)
        for ww in [win_w, start_w, tail_w]: ww.observe(refresh_feasibility, names="value")

        def render_preview(*_):
            with profile_preview_out:
                clear_output(wait=False)
                if profile_dd.value: display_wrapped_table(profile_preview(profile_dd.value), "Selected profile preview", max_rows=5)
        profile_dd.observe(render_preview, names="value")

        # INSTANT MODE SWITCHING (No lag, no clear_output)
        def render_mode(*_):
            if mode.value == "baseline":
                dynamic_panel.children = [ui_base]
            elif mode.value == "load":
                dynamic_panel.children = [ui_load]
                render_preview()
            else:
                dynamic_panel.children = [ui_edit]
                refresh_feasibility()

        mode.observe(render_mode, names="value")

        def _append_profile_option(path):
            opts = list(profile_dd.options) if hasattr(profile_dd, 'options') else []
            if str(path) not in opts: profile_dd.options = opts + [str(path)]
            profile_dd.value = str(path); mode.value = "load"

        def add_profile_path(_):
            with profile_msg:
                clear_output(wait=True)
                p = Path(profile_path_txt.value.strip())
                if not p.exists(): display(card("Not found", "Invalid path.", THEME["red"], THEME["red_bg"])); return
                try:
                    fp.load_user_method_profile(str(p))
                    _append_profile_option(str(p))
                    display(card("User profile accepted", "Review the preview, then click Activate.", THEME["green"], THEME["green_bg"]))
                except Exception as exc: display(card("Not accepted", f"<code>{exc}</code>", THEME["red"], THEME["red_bg"]))
        add_profile_btn.on_click(add_profile_path)

        def on_profile_upload(change):
            uploaded = normalize_upload_value(profile_upload.value)
            with profile_msg:
                clear_output(wait=True)
                if not uploaded: return
                meta = uploaded[0]
                target = Path(meta["name"]).name
                Path(target).write_bytes(bytes(meta["content"]))
                try:
                    fp.load_user_method_profile(target)
                    _append_profile_option(target)
                    display(card("Uploaded profile accepted", "Review preview, then click Activate.", THEME["green"], THEME["green_bg"]))
                except Exception as exc: display(card("Not accepted", f"<code>{exc}</code>", THEME["red"], THEME["red_bg"]))
        profile_upload.observe(on_profile_upload, names="value")

        def _changed_from_baseline(vals):
            return not (abs(vals["window_length_mm"]-25.0)<1e-12 and abs(vals["start_offset_mm"]-5.0)<1e-12 and abs(vals["tail_exclude_frac"]-0.20)<1e-12 and abs(vals["max_window_drift_frac"]-0.25)<1e-12 and abs(vals["peak_prominence_frac"]-0.003)<1e-12 and not vals["strict_extrema_spread_check"])

        def on_activate(_):
            global CONFIG, SETTINGS_CONFIRMED, ACTIVE_PROFILE_KIND, USER_PROFILE_PATH, ACTIVE_PROFILE_NOTE, PROFILE_CHANGED, RESULTS_STALE
            with out:
                clear_output(wait=False)
                base_cfg = make_baseline_config()
                if mode.value == "baseline":
                    CONFIG = base_cfg; ACTIVE_PROFILE_KIND = "MANUSCRIPT_BASELINE"; USER_PROFILE_PATH = ""; ACTIVE_PROFILE_NOTE = "Built-in manuscript_baseline_v1 selected."
                    display(final_banner("manuscript_baseline_v1 activated", "Controls are collapsed. Run Step 5 to analyze.<br>To change profile, rerun Step 4.", "ok"))
                elif mode.value == "load":
                    path = str(profile_dd.value or "")
                    if not path or not Path(path).exists():
                        display(card("No profile selected", "Select a JSON file.", THEME["red"], THEME["red_bg"])); return
                    profile = fp.load_user_method_profile(path)
                    CONFIG = fp.apply_user_profile_to_config(base_cfg, profile, allow_non_manuscript=True)
                    ACTIVE_PROFILE_KIND = "NON_MANUSCRIPT_MODIFIED_USER_PROFILE"; USER_PROFILE_PATH = path; ACTIVE_PROFILE_NOTE = str(profile.get("profile_note", ""))
                    display(final_banner("User profile activated", f"Loaded <code>{path}</code>.<br>Controls collapsed. Run Step 5 to analyze.", "warn"))
                else:
                    vals = {"window_length_mm": float(win_w.value), "start_offset_mm": float(start_w.value), "tail_exclude_frac": float(tail_w.value), "max_window_drift_frac": float(drift_w.value), "peak_prominence_frac": float(prom_w.value), "strict_extrema_spread_check": bool(spread_w.value)}
                    if not ack_w.value: display(card("Acknowledgment required", "Check the acknowledgment box.", THEME["red"], THEME["red_bg"])); return
                    if not _changed_from_baseline(vals): display(card("No settings changed", "These match the baseline. Change a setting or use baseline.", THEME["orange"], THEME["amber_bg"])); return
                    if vals["window_length_mm"] <= 0 or vals["start_offset_mm"] < 0 or not (0 <= vals["tail_exclude_frac"] < 0.8) or vals["max_window_drift_frac"] <= 0 or vals["peak_prominence_frac"] <= 0: return
                    _, affected = _window_feasibility_tables(vals["window_length_mm"], vals["start_offset_mm"], vals["tail_exclude_frac"])
                    if vals["window_length_mm"] < 12.5 and not short_window_ack_w.value: display(card("Required", "Check the short-window box.", THEME["red"], THEME["red_bg"])); return
                    if isinstance(affected, pd.DataFrame) and not affected.empty and not infeasible_window_ack_w.value: display(card("Required", "Check the feasibility acknowledgment.", THEME["red"], THEME["red_bg"])); return
                    CONFIG = base_cfg
                    for k,v in vals.items(): setattr(CONFIG, k, v)
                    CONFIG.allow_non_manuscript_run = True; CONFIG.method_profile = "non_manuscript_modified"; ACTIVE_PROFILE_KIND = "NON_MANUSCRIPT_MODIFIED_MANUAL_PROFILE"; ACTIVE_PROFILE_NOTE = profile_note_w.value.strip()
                    p_id = safe_profile_id(profile_id_w.value)
                    display_name = profile_display_w.value.strip() or p_id
                    profile_path = f"user_profile_{p_id}.json"
                    fp.save_method_profile_from_config(CONFIG, profile_path, p_id, display_name=display_name, note=ACTIVE_PROFILE_NOTE)
                    USER_PROFILE_PATH = profile_path
                    display(final_banner("User profile activated", f"Profile: <code>{p_id}</code><br>Controls collapsed. Run Step 5 to analyze.", "warn"))

                SETTINGS_CONFIRMED = True; PROFILE_CHANGED = True; RESULTS_STALE = True
                update_workflow_state(active_profile=ACTIVE_PROFILE_KIND, profile_note=ACTIVE_PROFILE_NOTE, results_stale=True)
                # INSTANT MENU COLLAPSE ON SUCCESS
                main_ui_box.layout.display = "none"

        btn_base.on_click(on_activate)
        btn_load.on_click(on_activate)
        btn_edit.on_click(on_activate)

        render_mode()
        display(widgets.VBox([main_ui_box, out]))
    else:
        CONFIG = make_baseline_config(); SETTINGS_CONFIRMED = True; ACTIVE_PROFILE_KIND = "MANUSCRIPT_BASELINE"; ACTIVE_PROFILE_NOTE = "Built-in manuscript_baseline_v1 selected."
        ok_card("manuscript_baseline_v1 selected", "Widgets unavailable. Run Step 5.")

step_timer_end("Step 4 profile selection", _t0)
```

```python
_t0 = step_timer_start("Step 7 reproducibility and output guide")
require_reference_validation()

#@title Step 7 â€” Reproducibility manager, reports, and what-next map
#@markdown Use this final step to open output files, save a user benchmark after a trusted run, and decide where to rerun safely.

def _outputs_available(obj):
    return isinstance(obj, dict) and bool(obj.get("outdir", ""))

def _outputs_summary_row(label, obj):
    if not _outputs_available(obj):
        return {
            "output_package": label,
            "available": False,
            "run_folder": "",
            "zip_file": "",
            "analysis_status": "",
            "metric_pass_count": "",
            "metric_fail_count": "",
            "manual_review_required_count": "",
        }

    readiness = obj.get("readiness_after_run", pd.DataFrame())
    row = readiness.iloc[0] if isinstance(readiness, pd.DataFrame) and not readiness.empty else {}

    return {
        "output_package": label,
        "available": True,
        "run_folder": obj.get("outdir", ""),
        "zip_file": obj.get("zip_file", ""),
        "analysis_status": row.get("analysis_status", ""),
        "metric_pass_count": row.get("metric_pass_count", ""),
        "metric_fail_count": row.get("metric_fail_count", ""),
        "manual_review_required_count": row.get("manual_review_required_count", ""),
    }

def _render_step7_package(active_outputs, active_label):
    """
    Render Step 7 using an explicitly selected output package.

    active_outputs may be:
      - OUTPUTS: main Step 5 manuscript/baseline or selected main-profile run
      - MODIFIED_OUTPUTS: optional Step 6 NON_MANUSCRIPT_MODIFIED recovery/sensitivity run
    """
    outdir = Path(active_outputs["outdir"])

    if active_label == "MODIFIED_OUTPUTS":
        display(
            card(
                "Step 7 is using MODIFIED_OUTPUTS",
                (
                    "You are reviewing the latest NON_MANUSCRIPT_MODIFIED recovery/sensitivity run. "
                    "This does not overwrite the main Step 5 OUTPUTS. Use this package for sensitivity/recovery review "
                    "unless you intentionally decide it is the package you want to benchmark or archive."
                ),
                THEME["orange"],
                THEME["amber_bg"],
            )
        )
    else:
        display(
            card(
                "Step 7 is using OUTPUTS",
                (
                    "You are reviewing the main Step 5 output package. This should be the manuscript-baseline package "
                    "when preparing manuscript-replica outputs."
                ),
                THEME["green"],
                THEME["green_bg"],
            )
        )

    display(
        card(
            "Output package ready",
            (
                "Default view is compact. Detailed audit files are saved in the selected run folder and ZIP. "
                "Start with <b>run_report.html</b> or <b>analysis_report.xlsx</b>, then review QC plots."
            ),
            "#37474f",
            "#f7f9fa",
        )
    )

    files_to_show = [
        ("Human-readable HTML report", active_outputs.get("run_report_html_file", "")),
        ("Markdown report", active_outputs.get("run_report_markdown_file", "")),
        ("Main ZIP", active_outputs.get("zip_file", "")),
        ("Excel report", active_outputs.get("excel_report_file", "")),
        ("Paper metrics CSV", active_outputs.get("paper_metrics_file", "")),
        ("Diagnostic summary CSV", active_outputs.get("diagnostic_summary_file", "")),
        ("Issue-action table CSV", active_outputs.get("issue_action_table_file", "")),
        ("Trace-integrity audit CSV", active_outputs.get("trace_integrity_audit_file", "")),
        ("Run history CSV", active_outputs.get("run_history_file", "")),
    ]

    display_wrapped_table(
        pd.DataFrame(
            [
                {
                    "file": lab,
                    "path": path,
                }
                for lab, path in files_to_show
                if path
            ]
        ),
        "Most useful output files",
        max_rows=20,
    )

    print("QC plots folder:", outdir / "qc_plots")

    if WIDGETS_AVAILABLE:
        bm_out = widgets.Output()

        open_bm = widgets.Button(
            description="Open user benchmark manager",
            button_style="",
            layout=widgets.Layout(width="280px"),
        )

        close_bm = widgets.Button(
            description="Close benchmark manager",
            layout=widgets.Layout(width="210px"),
        )

        def close_benchmark_manager(_):
            with bm_out:
                clear_output(wait=False)

        close_bm.on_click(close_benchmark_manager)

        def render_benchmark_manager(_):
            with bm_out:
                clear_output(wait=True)

                if active_label == "MODIFIED_OUTPUTS":
                    display(
                        card(
                            "Benchmark caution â€” modified output package selected",
                            (
                                "You are using a NON_MANUSCRIPT_MODIFIED output package. Create a user benchmark from this "
                                "only if you want a benchmark tied to this modified profile, not the manuscript baseline."
                            ),
                            THEME["orange"],
                            THEME["amber_bg"],
                        )
                    )

                display(
                    card(
                        "User benchmark manager",
                        (
                            "Use this optional section after a trusted run. A user benchmark checks whether the same "
                            "profile and data reproduce expected values in a future session. It does not replace or "
                            "overwrite the built-in Scotch reference validation."
                        ),
                        THEME["blue"],
                        THEME["pale_blue"],
                    )
                )

                display(close_bm)

                candidates = user_benchmark_candidates(".")

                if candidates:
                    display_wrapped_table(
                        pd.DataFrame({"detected_user_benchmark_expected_file": candidates}),
                        "Detected user benchmark files",
                        max_rows=10,
                    )

                    bench_dd = widgets.Dropdown(
                        options=candidates,
                        description="Validate:",
                        layout=widgets.Layout(width="95%"),
                    )

                    validate_btn = widgets.Button(
                        description="Validate selected output package against selected benchmark",
                        layout=widgets.Layout(width="460px"),
                    )

                    val_out = widgets.Output()

                    def on_validate(_):
                        with val_out:
                            clear_output(wait=True)

                            try:
                                vdf = fp.validate_user_benchmark(
                                    active_outputs.get("results", pd.DataFrame()),
                                    bench_dd.value,
                                )

                                display_wrapped_table(
                                    vdf.groupby("validation_status").size().reset_index(name="count"),
                                    "User benchmark validation counts",
                                    max_rows=10,
                                )

                                fails = (
                                    int((vdf.get("validation_status", pd.Series(dtype=str)) == "FAIL").sum())
                                    if not vdf.empty
                                    else 0
                                )

                                if fails:
                                    display(
                                        card(
                                            "User benchmark validation failed",
                                            (
                                                f"{fails} check(s) failed. This blocks only the selected "
                                                "user-benchmark association, not the built-in manuscript baseline "
                                                "if Step 1 passed."
                                            ),
                                            THEME["red"],
                                            THEME["red_bg"],
                                        )
                                    )
                                else:
                                    display(
                                        card(
                                            "User benchmark validation passed",
                                            "The selected output package reproduces the selected user benchmark within tolerance.",
                                            THEME["green"],
                                            THEME["green_bg"],
                                        )
                                    )

                            except Exception as exc:
                                display(
                                    card(
                                        "User benchmark validation error",
                                        f"<code>{type(exc).__name__}: {exc}</code>",
                                        THEME["red"],
                                        THEME["red_bg"],
                                    )
                                )

                    validate_btn.on_click(on_validate)

                    display(
                        widgets.VBox(
                            [
                                bench_dd,
                                widgets.HBox(
                                    [
                                        validate_btn,
                                        help_widget(
                                            "Validate against user benchmark",
                                            (
                                                "Compares the selected output package result table with the selected "
                                                "user-benchmark expected values. This validates only the user "
                                                "benchmark/profile association; it does not replace built-in reference validation."
                                            ),
                                            width="420px",
                                        ),
                                    ]
                                ),
                                val_out,
                            ]
                        )
                    )

                results = active_outputs.get("results", pd.DataFrame())
                sheets = (
                    list(results.get("sheet", pd.Series(dtype=str)).astype(str))
                    if isinstance(results, pd.DataFrame)
                    else []
                )

                if not sheets:
                    display(
                        card(
                            "No sheets available",
                            "No result rows are available for user-benchmark creation.",
                            THEME["orange"],
                            THEME["amber_bg"],
                        )
                    )
                    return

                bench_name = widgets.Text(
                    value="user_benchmark",
                    description="Name:",
                    layout=widgets.Layout(width="520px"),
                )

                bench_note = widgets.Textarea(
                    value=f"User benchmark created after QC review from {active_label}.",
                    description="Note:",
                    layout=widgets.Layout(width="95%", height="70px"),
                )

                bench_sheets = widgets.SelectMultiple(
                    options=sheets,
                    value=tuple(sheets[: min(3, len(sheets))]),
                    description="Sheets:",
                    layout=widgets.Layout(width="95%", height="140px"),
                )

                ack = widgets.Checkbox(
                    value=False,
                    description=(
                        "I reviewed QC plots, metric-quality flags, and export integrity, and I want to save "
                        "selected results from the selected output package as a user benchmark."
                    ),
                    layout=widgets.Layout(width="95%"),
                )

                modified_ack = widgets.Checkbox(
                    value=False,
                    description=(
                        "If using MODIFIED_OUTPUTS, I understand this benchmark belongs to the modified profile, "
                        "not the manuscript baseline."
                    ),
                    layout=widgets.Layout(width="95%"),
                )

                save_bench = widgets.Button(
                    description="Save selected rows as user benchmark",
                    button_style="primary",
                    layout=widgets.Layout(width="340px"),
                )

                status = widgets.Output()

                def on_save(_):
                    with status:
                        clear_output(wait=True)

                        if not ack.value:
                            display(
                                card(
                                    "Acknowledgment required",
                                    (
                                        "Confirm that you reviewed QC plots and metric-quality flags before "
                                        "creating a user benchmark."
                                    ),
                                    THEME["red"],
                                    THEME["red_bg"],
                                )
                            )
                            return

                        if active_label == "MODIFIED_OUTPUTS" and not modified_ack.value:
                            display(
                                card(
                                    "Modified-profile acknowledgment required",
                                    (
                                        "Because MODIFIED_OUTPUTS is selected, confirm that this benchmark belongs "
                                        "to the modified profile rather than the manuscript baseline."
                                    ),
                                    THEME["red"],
                                    THEME["red_bg"],
                                )
                            )
                            return

                        name = bench_name.value.strip() or "user_benchmark"
                        expected = f"{name}_expected.csv"
                        meta = f"{name}_metadata.json"

                        try:
                            paths = fp.create_user_benchmark_from_results(
                                active_outputs["results"],
                                expected_path=expected,
                                metadata_path=meta,
                                selected_sheets=list(bench_sheets.value),
                                note=bench_note.value,
                            )

                            display(
                                card(
                                    "User benchmark saved",
                                    (
                                        f"Expected values: <code>{paths['expected_path']}</code><br>"
                                        f"Metadata: <code>{paths['metadata_path']}</code><br>"
                                        "Use these files with the associated user profile in future sessions. "
                                        "Do not overwrite the built-in Scotch reference."
                                    ),
                                    THEME["green"],
                                    THEME["green_bg"],
                                )
                            )

                        except Exception as exc:
                            display(
                                card(
                                    "User benchmark not saved",
                                    f"<code>{type(exc).__name__}: {exc}</code>",
                                    THEME["red"],
                                    THEME["red_bg"],
                                )
                            )

                save_bench.on_click(on_save)

                children = [
                    bench_name,
                    bench_note,
                    bench_sheets,
                    ack,
                ]

                if active_label == "MODIFIED_OUTPUTS":
                    children.append(modified_ack)

                children.extend(
                    [
                        widgets.HBox(
                            [
                                save_bench,
                                help_widget(
                                    "Create user benchmark",
                                    (
                                        "Saves selected rows from the selected trusted output package as your own "
                                        "benchmark files. Use only after reviewing QC plots, metric-quality flags, "
                                        "and export integrity."
                                    ),
                                    width="420px",
                                ),
                            ]
                        ),
                        status,
                    ]
                )

                display(widgets.VBox(children))

        open_bm.on_click(render_benchmark_manager)

        display(
            widgets.VBox(
                [
                    widgets.HBox(
                        [
                            open_bm,
                            help_widget(
                                "User benchmark manager",
                                (
                                    "Open this only if you want to create or validate a separate user benchmark "
                                    "from the selected trusted output package. This is useful for future "
                                    "reproducibility checks with your own lab data."
                                ),
                                width="380px",
                            ),
                        ]
                    ),
                    bm_out,
                ]
            )
        )

    show_detail_button_with_help(
        "Show full output file guide",
        lambda: fp.output_file_guide(),
        "Show full output file guide",
        (
            "Opens the full output-file dictionary only when needed. Start with run_report.html, "
            "analysis_report.xlsx, paper_metrics.csv, group_summary.csv, and qc_plots/."
        ),
    )

    display(
        card(
            "What next?",
            "<b>Analyze a new file:</b> Step 2 â†’ clear current file â†’ select/upload new file.<br>"
            "<b>Change method/profile:</b> Step 4 â†’ activate manuscript_baseline_v1 or a user profile; rerun Step 5.<br>"
            "<b>Change columns, units, geometry, labels, notes, or included sheets:</b> Step 3 â†’ update mapping â†’ confirm â†’ rerun Step 5.<br>"
            "<b>Run recovery/sensitivity:</b> Step 6 after reviewing diagnostics.<br>"
            "<b>Review recovery/sensitivity outputs:</b> Step 7 â†’ select MODIFIED_OUTPUTS in the output selector.<br>"
            "<b>Create a reusable user profile:</b> Step 3 or Step 6; download the JSON if using Binder/Colab.<br>"
            "<b>Create a user benchmark:</b> open the manager above after a trusted selected run; never overwrite bundled Scotch reference files.<br>"
            "<b>Use manuscript-replica outputs:</b> use only runs labeled MANUSCRIPT_BASELINE with reference validation PASS, export integrity PASS, and QC plots reviewed.",
            THEME["blue"],
            THEME["pale_blue"],
        )
    )

    if RUN_HISTORY:
        display_wrapped_table(
            pd.DataFrame(RUN_HISTORY),
            "Final run comparison",
            max_rows=30,
        )


main_outputs = globals().get("OUTPUTS", None)
modified_outputs = globals().get("MODIFIED_OUTPUTS", None)

main_available = _outputs_available(main_outputs)
modified_available = _outputs_available(modified_outputs)

if not main_available and not modified_available:
    waiting(
        "Output/reproducibility tools waiting for analysis",
        "Run Step 5 first. Optional Step 6 modified outputs can also be reviewed here after Step 5.",
    )

else:
    summary_df = pd.DataFrame(
        [
            _outputs_summary_row("OUTPUTS â€” main Step 5 package", main_outputs),
            _outputs_summary_row("MODIFIED_OUTPUTS â€” latest Step 6 recovery/sensitivity package", modified_outputs),
        ]
    )

    display_wrapped_table(
        summary_df,
        "Available output packages",
        max_rows=5,
    )

    if WIDGETS_AVAILABLE:
        selector_out = widgets.Output()

        options = []

        if main_available:
            options.append(("Use OUTPUTS â€” main Step 5 package", "main"))

        if modified_available:
            options.append(("Use MODIFIED_OUTPUTS â€” latest Step 6 package", "modified"))

        default_key = globals().get("STEP7_ACTIVE_OUTPUT_KEY", "main")

        allowed_keys = [v for _, v in options]

        if default_key not in allowed_keys:
            default_key = allowed_keys[0]

        output_choice = widgets.Dropdown(
            options=options,
            value=default_key,
            description="Step 7 uses:",
            layout=widgets.Layout(width="95%"),
        )

        render_btn = widgets.Button(
            description="Render selected output package",
            button_style="primary",
            layout=widgets.Layout(width="300px"),
        )

        def render_selected_output(_=None):
            globals()["STEP7_ACTIVE_OUTPUT_KEY"] = output_choice.value

            with selector_out:
                clear_output(wait=True)

                if output_choice.value == "modified":
                    _render_step7_package(modified_outputs, "MODIFIED_OUTPUTS")
                else:
                    _render_step7_package(main_outputs, "OUTPUTS")

        render_btn.on_click(render_selected_output)

        selector_intro = card(
            "Choose which output package Step 7 should display",
            (
                "This selector does not overwrite OUTPUTS or MODIFIED_OUTPUTS. It only controls which "
                "package Step 7 displays and uses for optional user-benchmark actions."
            ),
            THEME["blue"],
            THEME["pale_blue"],
        )

        display(selector_intro)

        display(
            widgets.VBox(
                [
                    widgets.HBox(
                        [
                            output_choice,
                            render_btn,
                            help_widget(
                                "Step 7 output selector",
                                (
                                    "Use OUTPUTS for manuscript-baseline/main results. Use MODIFIED_OUTPUTS only "
                                    "when intentionally reviewing a Step 6 recovery/sensitivity run."
                                ),
                                width="390px",
                            ),
                        ]
                    ),
                    selector_out,
                ]
            )
        )

        render_selected_output()

    else:
        if main_available:
            display(
                card(
                    "Widgets unavailable â€” using OUTPUTS",
                    "Step 7 is displaying the main Step 5 output package.",
                    THEME["green"],
                    THEME["green_bg"],
                )
            )
            _render_step7_package(main_outputs, "OUTPUTS")
        else:
            display(
                card(
                    "Widgets unavailable â€” using MODIFIED_OUTPUTS",
                    "Step 7 is displaying the latest Step 6 modified output package because no main OUTPUTS package is available.",
                    THEME["orange"],
                    THEME["amber_bg"],
                )
            )
            _render_step7_package(modified_outputs, "MODIFIED_OUTPUTS")

step_timer_end("Step 7 reproducibility and output guide", _t0)
```

## Interpretation notes

- `SSA` is an operational force-oscillation descriptor under the same machine/protocol, not an intrinsic crack-front material property.
- `PSI` is only meaningful when the selected window is physically plausible and passes the drift/tail/window checks.
- `MANUSCRIPT_BASELINE` outputs reproduce the locked analysis settings. `NON_MANUSCRIPT_MODIFIED` outputs are recovery/sensitivity runs and should not be mixed with manuscript-baseline results without explicit disclosure.
- `displacement_at_break_mm` uses a post-window `0.10 Ã— Fc` drop proxy when found; if no drop is detected, the final recorded displacement is used and flagged.
- QC plots use color-blind-safe, paper-matched colors, but interpretation should rely on both color and marker/line style plus the exported diagnostic tables.
